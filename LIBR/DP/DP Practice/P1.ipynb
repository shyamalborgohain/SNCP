## 416. Partition Equal Subset Sum (LeetCode)

```python
class Solution:
    def canPartition(self, nums):
        total = sum(nums)

        # Odd total cannot be divided into two equal subsets
        if total % 2 != 0:
            return False

        target = total // 2

        dp = [False] * (target + 1)
        dp[0] = True

        for num in nums:
            # Traverse from backward, so each number is used only once
            for j in range(target, num - 1, -1):
                dp[j] = dp[j] or dp[j - num]

        return dp[target]
```

```python
nums = [1, 5, 11, 5]

print(Solution().canPartition(nums))
```

## Partition Into 2 Subsets with Min Sum Diff (GeeksforGeeks)

```python
class Solution:
    def minDifference(self, arr):
        total = sum(arr); target = total//2; dp = [False]*(target+1); dp[0] =True
        for i in arr:
            for j in range(target, i-1, -1): dp[j] = dp[j] or dp[j-i]
        while not dp[target]: target -= 1
        return total - 2*target
```

```python
nums = [1, 5, 11, 5]

print(Solution().minDifference(nums))
```

## Count Subsets with Sum Target (GeeksforGeeks)

```python
class Solution:
    def perfectSum(self, arr, target):
        dp = [0]*(target+1); dp[0] = 1
        for a in arr:
            for j in range(target, a-1, -1): dp[j] += dp[j-a]
        return dp[target]
```

```python
arr = [5, 2, 3, 10, 6, 8]
target = 10
print(Solution().perfectSum(arr, target))
```

## Count Partitions (GeeksforGeeks), Equivalent to Target Sum Problem

```python
class Solution:
    def countPartitions(self, arr, diff):
        total = sum(arr)
        if diff > total or (total - diff)%2 != 0: return 0
        target = (total - diff)//2; dp = [0]*(target+1); dp[0] = 1
        for a in arr:
            for j in range(target, a-1, -1): dp[j] += dp[j-a]
        return dp[-1]
```

```python
arr = [5, 2, 6, 4]
diff = 3
print(Solution().countPartitions(arr, diff))
```

## 0-1 Knapsack Problem (GeeksforGeeks)

```python
class Solution:
    def knapsack(self, W: int, val: list[int], wt: list[int]) -> int:
        n = len(wt); dp = [0]*(W+1)
        for i in range(n):
            for j in range(W, wt[i]-1, -1): dp[j] = max(dp[j], val[i] + dp[j-wt[i]])
        return dp[-1]
```

```python
W = 5
val = [10, 40, 30, 50]
wt = [5, 4, 2, 3] 
print(Solution().knapsack(W, val, wt))
```

## Unbounded Knapsack (GeeksforGeeks)

```python
class Solution:
    def knapSack(self, val, wt, capacity):
        n = len(val); dp = [0]*(capacity+1);
        for i in range(n):
            for j in range(wt[i], capacity+1): dp[j] = max(dp[j-wt[i]] + val[i], dp[j])
        return dp[-1]
```

```python
val = [10, 40, 50, 70] 
wt = [1, 3, 4, 5]
capacity = 8
print(Solution().knapSack(val, wt, capacity))
```

## Minimum Coins Change (GeeksforGeeks)

```python
class Solution:
    def minCoins(self, coins: list[int], sum: int) -> int:
        dp = [10**5]*(sum+1); dp[0] = 0
        for coin in coins:
            for j in range(coin, sum+1): dp[j] = min(dp[j-coin]+1, dp[j])
        return -1 if dp[-1] == 10**5 else dp[-1]
```

```python
coins = [9, 6, 5, 1]
sum = 19
print(Solution().minCoins(coins, sum))
```

## Coin Change (Count Ways) (GeeksforGeeks)

```python
class Solution:
    def count(self, coins: list[int], sm: int) -> int:
        dp = [0]*(sm+1); dp[0] = 1
        for x in coins:
            for j in range(x, sm+1): dp[j] += dp[j-x]
        return dp[-1]
```

```python
coins = [2, 5, 3, 6]
sm = 10
print(Solution().count(coins, sm))
```

## WildCard Pattern Matching (GeeksforGeeks)

```python
class Solution:
    def wildCard(self, txt: str, pat: str) -> bool:
        n, m = len(txt), len(pat); dp = [[False]*(m+1)for _ in range(n+1)]; dp[0][0] = True
        for i in range(1, m+1): 
            if pat[i-1] == '*': dp[0][i] = dp[0][i-1]
        for i in range(1, n+1):
            for j in range(1, m+1):
                if pat[j-1] == '*': dp[i][j] = dp[i-1][j] or dp[i][j-1]
                elif pat[j-1] == '?' or pat[j-1] == txt[i-1]: dp[i][j] = dp[i-1][j-1]
                else: dp[i][j] = False
        return dp[-1][-1]
```

```python
txt = "baaabab"
pat = "a*ab"
print(Solution().wildCard(txt, pat))
```

## 123. Best Time to Buy and Sell Stock III (LeetCode)

```python
class Solution:
    def maxProfit(self, prices: list[int]) -> int:
        buy1 = buy2 = float('inf'); sell1 = sell2 = 0
        for price in prices:
            buy1 = min(buy1, price); sell1 = max(sell1, price - buy1)
            buy2 = min(buy2, price - sell1); sell2 = max(sell2, price - buy2)
        return sell2
```

```python
prices = [3,3,5,0,0,3,1,4]
print(Solution().maxProfit(prices))
```

## Longest String Chain

```python
class Solution:
    def longestStringChain(self, words):
        words.sort(key=len); dp = {}; ans = 0
        for word in words:
            dp[word] = 1
            for i in range(len(word)):
                prev = word[:i] + word[i+1:]
                if prev in dp: dp[word] = max(dp[prev]+1, dp[word])
            ans = max(ans, dp[word])
            # if ans >= 10: return ans
        return ans
```

```python
words = ["ba", "b", "a", "bca", "bda", "bdca"]
print(Solution().longestStringChain(words))
```

## Matrix Chain Multiplication (GeeksforGeeks)

```python
class Solution:
    def matrixMultiplication(self, arr):
        n = len(arr); dp = [[0]*n for _ in range(n)]
        for ln in range(2, n):
            for i in range(n-ln):
                j = i+ln; dp[i][j] = 10**11
                for k in range(i+1, j):
                    dp[i][j] = min(dp[i][j], dp[i][k] + dp[k][j] + arr[i]*arr[j]*arr[k])
        return dp[0][-1]
```

```python
arr = [2, 1, 3, 4]
print(Solution().matrixMultiplication(arr))
```

## Minimum Cost to cut a Stick of length N (GeeksforGeeks)

```python
class Solution:
    def minCutCost(self, n, cuts):
        cuts.sort(); cuts = [0]+cuts+[n]; m = len(cuts); dp = [[0]*m for _ in range(m)]
        for ln in range(2, m):
            for i in range(m-ln):
                j = i+ln; dp[i][j] = 10**23
                for k in range(i+1, j):
                    cost = cuts[j]-cuts[i]+dp[i][k]+dp[k][j]; dp[i][j] = min(dp[i][j], cost)
        return dp[0][-1]
```

```python
n = 8; cuts = [1, 5, 3, 6]
print(Solution().minCutCost(n, cuts))
```

## Boolean Parenthesization (GeeksforGeeks)

```python
class Solution:
    def countWays(self, s):
        n = len(s); T = [[0]*n for _ in range(n)]; F = [[0]*n for _ in range(n)]
        for i in range(0, n, 2):
            if s[i] == 'T': T[i][i] = 1
            else: F[i][i] = 1
        for ln in range(2, n, 2):
            for i in range(0, n-ln, 2):
                j = i+ln
                for k in range(i+1, j, 2):
                    lt, lf = T[i][k-1], F[i][k-1]; rt, rf = T[k+1][j], F[k+1][j]
                    if s[k] == '&': T[i][j] += lt*rt; F[i][j] += lt*rf + lf*rt + lf*rf
                    elif s[k] == '|': T[i][j] += lt*rf + lt*rt + lf*rt; F[i][j] += lf*rf
                    else: T[i][j] += lf*rt + lt*rf; F[i][j] += lf*rf + lt*rt
        return T[0][-1]
```

```python
s = "T|T&F^T"
print(Solution().countWays(s))
```

## Maximum Partition Sum

```python
class Solution:
    def maxPartitionSum(self, arr, k):
        n = len(arr); dp = [0]*(n+1)
        for i in range(1, n+1):
            mx = 0
            for length in range(1, min(k, i)+1):
                mx = max(mx, arr[i - length])
                dp[i] = max(dp[i], dp[i-length] + mx*length)
        return dp[n]
```

```python
arr =[4, 6, 1, 1, 1, 1]
print(Solution().maxPartitionSum(arr, 2))
```

## Count Square Submatrices with all 1's

```python
class Solution:
    def countSquares(self, mat):
        n = len(mat); m = len(mat[0]); dp = [[0]*m for _ in range(n)]; ans = 0
        for i in range(n):
            for j in range(m):
                if mat[i][j] == 1:
                    if i == 0 or j == 0: dp[i][j] = 1
                    else: dp[i][j] = 1 + min(dp[i-1][j-1], dp[i-1][j], dp[i][j-1])
                    ans += dp[i][j]
        return ans
```

```python
n = 3; m = 3; mat = [[0, 1, 1], [1, 1, 1], [0, 1, 1]]
print(Solution().countSquares(mat))
```

# Chocolate Pickup(GeeksforGeeks)

### Using 3D DP, Time Complexity: $O(nm^2)$ and Space Complexity $O(nm^2)$

```python
class Solution:
    def maxChocolates1(self, grid):
        n = len(grid); m = len(grid[0]); dp = [[[-1]*m for _ in range(m)] for _ in range(n)]
        # Base case: last row
        for c1 in range(m):
            for c2 in range(m):
                if c1 == c2: dp[n-1][c1][c2] = grid[n-1][c1]
                else: dp[n-1][c1][c2] = (grid[n-1][c1] + grid[n-1][c2])
        # Move from bottom to top
        for r in range(n-2, -1, -1):
            for c1 in range(m):
                for c2 in range(m):
                    best = -10**9
                    for d1 in [-1, 0, 1]:
                        for d2 in [-1, 0, 1]:
                            nc1 = c1 + d1; nc2 = c2 + d2
                            if 0 <= nc1 < m and 0 <= nc2 < m:
                                best = max(best,dp[r + 1][nc1][nc2])
                    curr = grid[r][c1]
                    if c1 == c2: curr += best
                    else: curr += grid[r][c2] + best
                    dp[r][c1][c2] = curr
        return dp[0][0][m - 1]
```

```python
grid = [[4, 1, 2], [3, 6, 1], [1, 6, 6], [3, 1, 2]]
print(Solution().maxChocolates1(grid))
```

### Using 2D DP, Time Complexity: $O(nm^2)$ and Space Complexity $O(nm)$, slightly faster than 3D DP due to lesser use of memory

```python
class Solution:
    def maxChocolates2(self, grid):
        n = len(grid); m = len(grid[0]); dp = [[0]*m for _ in range(m)]
        # Best Case: Last row
        for c1 in range(m):
            for c2 in range(m):
                dp[c1][c2] = grid[n-1][c1]
                if c1 != c2: dp[c1][c2] += grid[n-1][c2]
        for r in range(n - 2, -1, -1):
            new = [[-10**9]*m for _ in range(m)]
            for c1 in range(m):
                for c2 in range(m):
                    for d1 in (-1, 0, 1):
                        for d2 in (-1, 0, 1):
                            nc1 = c1 + d1; nc2 = c2 + d2
                            if 0 <= nc1 < m and 0 <= nc2 < m:
                                curr = grid[r][c1]
                                if c1 != c2: curr += grid[r][c2]
                                new[c1][c2] = max(new[c1][c2], curr + dp[nc1][nc2])
            dp = new
        return dp[0][m - 1]
```

```python
grid = [[4, 1, 2], [3, 6, 1], [1, 6, 6], [3, 1, 2]]
print(Solution().maxChocolates2(grid))
```
