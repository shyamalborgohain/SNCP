# **DATA STRUCTURES**

## Array(List)

- An ordered and mutable collection of elements.
- Dynamic size
- Allows duplicates
- Maintains insertion order
- Supports indexing
- Random access

| Operation | Access | Search | Append | Insert | Delete |
|-----------|--------|--------|--------|--------|--------|
| Time  Complexity | $O(1)$ | $O(n)$ | $O(1)$ | $O(n)$ | $O(n)$ |

- Most commonly used data structure in Python.
- Implemented as a dynamic array.
- Best when frequent indexing is required.

````python
arr = [1, 2, 3]

arr.append(4)
arr.insert(1, 10)

print(arr[0])

arr.pop()
````

````python
arr = [10, 20, 30, 40, 50]

mid = len(arr) // 2

arr[:]      # copy
arr[::-1]   # reverse
arr[:mid]   # left half
arr[mid:]   # right half
arr[-2:]    # last 2 elements
arr[:-2]    # remove last 2 elements
arr[::2]    # even indices
arr[1::2]   # odd indices
````

## Tuple

- An ordered, immutable collection of elements.
- Allows duplicates
- Supports indexing and slicing
- Hashable (if elements are hashable)

## Complexity

| Operation | Access | Search | Count | Index |
|------------|------|---------|-------|-------|
| Time Complexity | $O(1)$ | $O(n)$ | $O(n)$ | $O(n)$ |

- `count(x)`
- `index(x)`
- Fixed data
- Returning multiple values
- Dictionary keys

````python
# Tuple
t = (1, 2, 3)

# Access
print(t[0])

# Unpacking
a, b, c = t

# Methods
print(t.count(2))
print(t.index(3))

# Tuple as dictionary key
d = {(1, 2): "point"}
````

## Set

- An unordered collection of unique elements.
- No duplicate values
- Fast lookup
- Mutable
- Hash-based implementation

| Operation | Add | Remove | Search |
|------------|------|------|--------|
| Time Complexity | $O(1)$ | $O(1)$ | $O(1)$ |

- Useful for removing duplicates.
- Best for membership testing.
- Elements must be hashable.

````python
s = {1, 2, 3}

s.add(4)
s.remove(2)

print(3 in s)
````

## Dictionary

- A collection of key-value pairs.
- Unique keys
- Fast lookup
- Mutable
- Maintains insertion order

| Operation | Insert | Delete | Search |
|------------|------|---------|--------|
| Time Complexity | $O(1)$ | $O(1)$ | $O(1)$ |

- Implemented using a hash table.
- Keys must be hashable.
- Most used data structure in DSA problems.

````python
# Example input
nums = [1, 2, 2, 3, 3, 3]

# Frequency Map using dict
freq = {}
for x in nums:
    freq[x] = freq.get(x, 0) + 1

print(freq)
# {1: 1, 2: 2, 3: 3}
````

````python
from collections import defaultdict

nums = [1, 2, 2, 3, 3, 3]

# Frequency Map using defaultdict
freq = defaultdict(int)

for x in nums:
    freq[x] += 1

print(freq)
# defaultdict(<class 'int'>, {1: 1, 2: 2, 3: 3})
````

````python
from collections import Counter

nums = [1, 2, 2, 3, 3, 3]

# Frequency Map using Counter
freq = Counter(nums)

print(freq)
# Counter({3: 3, 2: 2, 1: 1})
````

````python
# Normal dictionary
d = {}

# Missing key -> 0
defaultdict(int)

# Missing key -> []
defaultdict(list)

# Missing key -> set()
defaultdict(set)

# Frequency count
Counter(nums)

# Ordered dictionary
OrderedDict()

# Read-only dictionary
MappingProxyType(d)

# Combine dictionaries
ChainMap(d1, d2)
````

## Stack

- A linear data structure following LIFO (Last In First Out).
- Insert and delete from the same end.
- Fast push and pop operations.
- Used in recursion and DFS.

| Operation | Push | Pop | Top |
|------------|------|----|-----|
| Time Complexity | $O(1)$ | $O(1)$ | $O(1)$ |

- Can be implemented using a Python list.
- Last inserted element is removed first.

````python
# Stack (using list)

stack = []                         # O(1)

# Push                            # O(1)
stack.append(10)
stack.append(20)
stack.append(30)

# [10, 20, 30]

# Top / Peek                      # O(1)
stack[-1]

# Pop Top                         # O(1)
stack.pop()

# [10, 20]

# Pop by Index                    # O(n)
stack.pop(0)       # first
stack.pop(1)       # index 1
stack.pop(-1)      # last

# Insert                          # O(n)
stack.insert(0, 5)
stack.insert(1, 15)

# Remove by Value                 # O(n)
stack.remove(20)

# Search                          # O(n)
20 in stack

# Count                           # O(n)
stack.count(20)

# Index                           # O(n)
stack.index(20)

# Extend                          # O(k)
stack.extend([40, 50])

# Reverse                         # O(n)
stack.reverse()

# Sort                            # O(n log n)
stack.sort()

# Copy                            # O(n)
stack.copy()
stack[:]

# Clear                           # O(n)
stack.clear()

# Size                            # O(1)
len(stack)

# Empty                           # O(1)
not stack

# Access                          # O(1)
stack[0]
stack[-1]
stack[1]

# Slicing                         # O(k)
stack[:]
stack[1:]
stack[:-1]
stack[::-1]

# Standard Stack Operations

# Push           O(1)
# Pop            O(1)
# Top / Peek     O(1)
# Size           O(1)
# Empty Check    O(1)

# Search         O(n)
# Traverse       O(n)
````

````python
stack.append(x)   # Push
stack.pop()       # Pop
stack[-1]         # Top
len(stack)
not stack
````

# Queue

- A linear data structure following FIFO (First In First Out).
- Insert from rear.
- Delete from front.
- Used in BFS and scheduling.

| Operation | Enqueue | Dequeue |
|------------|------|-----------|
| Time Complexity | $O(1)$ | $O(1)$ |

- Implement using ```collections.deque```
- First inserted element is removed first.

````python
from collections import deque

# Create
q = deque()                      # O(1)
q = deque([10, 20, 30])          # O(n)

# Enqueue
q.append(40)                     # O(1)

# Dequeue
q.popleft()                      # O(1)

# Front
q[0]                             # O(1)

# Rear
q[-1]                            # O(1)

# Size
len(q)                           # O(1)

# Empty
not q                            # O(1)

# Extend
q.extend([40, 50])               # O(k)

# Search
q.count(20)                      # O(n)
q.index(20)                      # O(n)

# Remove by value
q.remove(20)                     # O(n)

# Reverse
q.reverse()                      # O(n)

# Rotate
q.rotate(1)                      # O(k)
q.rotate(-1)                     # O(k)

# Clear
q.clear()                        # O(n)
````

````python
# Queue (FIFO)

q = deque()

q.append(10)     # Enqueue
q.append(20)
q.append(30)

q.popleft()      # Dequeue -> 10

# Front
q[0]

# Rear
q[-1]
````

````python
from collections import deque

q = deque()

q.append(x)      # O(1)
q.popleft()      # O(1)

q[0]             # O(1)
q[-1]            # O(1)

len(q)           # O(1)
not q            # O(1)
````

# Deque

- A double-ended queue supporting insertion and deletion from both ends.
- Works as both stack and queue.
- Efficient operations at both ends.

| Operation | Append Left | Append Right | Pop Left | Pop Right |
|------------|------|-------|-----|------|
| Time Complexity | $O(1)$ | $O(1)$ | $O(1)$ | $O(1)$ |

- Implemented using ```collections.deque```
- Preferred over list for queue operations.

````python
from collections import deque

# Create
dq = deque()                     # O(1)
dq = deque([10, 20, 30])         # O(n)

# Add
dq.append(40)                    # O(1)
dq.appendleft(5)                 # O(1)

# Remove
dq.pop()                         # O(1)
dq.popleft()                     # O(1)

# Access
dq[0]                            # O(1)
dq[-1]                           # O(1)

# Extend
dq.extend([40, 50])              # O(k)
dq.extendleft([5, 0])            # O(k)

# Insert
dq.insert(1, 15)                 # O(n)

# Remove by value
dq.remove(20)                    # O(n)

# Search
dq.count(20)                     # O(n)
dq.index(20)                     # O(n)

# Reverse
dq.reverse()                     # O(n)

# Rotate
dq.rotate(1)                     # O(k)
dq.rotate(-1)                    # O(k)

# Clear
dq.clear()                       # O(n)

# Size
len(dq)                          # O(1)

# Empty
not dq                           # O(1)

# Queue
dq.append(x)                     # O(1)
dq.popleft()                     # O(1)

# Stack
dq.append(x)                     # O(1)
dq.pop()                         # O(1)
````

````python
# Most used in DSA

from collections import deque

dq = deque()

dq.append(x)
dq.appendleft(x)

dq.pop()
dq.popleft()

dq[0]
dq[-1]

len(dq)
not dq
````

# Heap

- A complete binary tree satisfying the heap property.
- Python provides Min Heap.
- Efficient priority operations.
- Stored internally as an array.

| Operation | Insert | Extract Min | Peek |
|------------|------|------|-------|
| Time Complexity | $O(\log n)$ | $O(\log n)$ | $O(1)$ |

- Used for Priority Queue problems.
- Root always contains the minimum element.

````python
import heapq

# Create
heap = []                         # O(1)

# Heapify
heap = [4, 1, 3, 2]
heapq.heapify(heap)               # O(n)

# Insert
heapq.heappush(heap, 5)           # O(log n)

# Remove smallest
heapq.heappop(heap)               # O(log n)

# Smallest element
heap[0]                           # O(1)

# Push then pop
heapq.heappushpop(heap, 5)        # O(log n)

# Pop then push
heapq.heapreplace(heap, 5)        # O(log n)

# n smallest
heapq.nsmallest(3, heap)          # O(n log k)

# n largest
heapq.nlargest(3, heap)           # O(n log k)

# Merge sorted iterables
heapq.merge(a, b)                 # O(n)

# Size
len(heap)                         # O(1)

# Empty
not heap                          # O(1)
````

# Linked List

- A collection of nodes where each node stores data and a reference to the next node.
- Dynamic size.
- Non-contiguous memory.
- Efficient insertion and deletion.
- Sequential access.

| Operation | Access | Search | Insert Head | Delete Head |
|------------|------|----|-----|-----|
| Time Complexity | $O(n)$ | $O(n)$ | $O(1)$ | $O(1)$ |

- No direct indexing.
- Better than arrays for frequent insertions/deletions.
- Common in interview questions.
- Types - Singly Linked List, Doubly Linked List, Circular Linked List

````python
class Node:
    def __init__(self, val):
        self.val = val
        self.next = None


head = Node(1)
head.next = Node(2)
head.next.next = Node(3)
````

````python
# Singly Linked List

class Node:
    def __init__(self, val):
        self.val = val
        self.next = None

# Create
head = Node(10)
head.next = Node(20)
head.next.next = Node(30)

# Traverse                      # O(n)
curr = head
while curr:
    print(curr.val)
    curr = curr.next

# Insert at Head                # O(1)
new = Node(5)
new.next = head
head = new

# Insert at Tail                # O(n)
new = Node(40)
curr = head
while curr.next:
    curr = curr.next
curr.next = new

# Delete Head                   # O(1)
head = head.next

# Search                        # O(n)
target = 20
curr = head
while curr:
    if curr.val == target:
        break
    curr = curr.next

# Length                        # O(n)
count = 0
curr = head
while curr:
    count += 1
    curr = curr.next

# Reverse                       # O(n)
prev = None
curr = head

while curr:
    nxt = curr.next
    curr.next = prev
    prev = curr
    curr = nxt

head = prev

# Middle Node                   # O(n)
slow = fast = head

while fast and fast.next:
    slow = slow.next
    fast = fast.next.next

# Cycle Detection               # O(n)
slow = fast = head

while fast and fast.next:
    slow = slow.next
    fast = fast.next.next

    if slow == fast:
        break
````

````python
# Standard Complexities

# Access          O(n)
# Search          O(n)
# Traverse        O(n)

# Insert Head     O(1)
# Delete Head     O(1)

# Insert Tail     O(n)
# Delete Tail     O(n)

# Reverse         O(n)
````
