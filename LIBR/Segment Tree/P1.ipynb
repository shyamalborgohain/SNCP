## SEGMENT TREE

### What is a Segment Tree?

A **Segment Tree** is a binary tree used to efficiently answer **range queries** and perform **point updates** on an array.

Examples of range queries:
- Range Sum
- Range Minimum
- Range Maximum
- Range GCD
- Range XOR
- Counting Frequencies
- Lazy Propagation(Range Updates)

---

## Why use a Segment Tree?

Given an array:

```text
[5, 2, 7, 1]
```

If we need the sum from index `1` to `3`:

- Brute Force: **O(n)**
- Segment Tree: **O(log n)**

For many queries and updates, a Segment Tree is much faster.

---

## Structure

Each node stores information about a **segment (range)** of the array.

Example:

```text
Array

Index : 0 1 2 3
Value : 5 2 7 1
```

Segment Tree (Sum):

```text
                [0..3] = 15
               /            \
        [0..1] = 7       [2..3] = 8
          /     \          /      \
      [0]=5   [1]=2    [2]=7   [3]=1
```

---

## Terminology

- Root → Whole array
- Internal Node → Segment (range)
- Leaf Node → Single element

---

## Complexity

| Operation | Build | Query | Update | Memory |
|-----------|-------|-------|--------|--------|
| Time | O(n) | O(log n) | O(log n) | O(4n) |

---

## Array Representation

Instead of using tree nodes, we use an array.

```text
tree[1] → Root
tree[2] → Left Child
tree[3] → Right Child
```

Relationships:

```text
Left Child  = 2*node
Right Child = 2*node + 1
```

Memory:

```python
tree = [0]*(4*n)
```

---

## Query Cases

Suppose we query `[2,5]`.

### 1. No Overlap

```text
Query : 2------5
Node  :         6----8
```

Ignore this node.

---

### 2. Complete Overlap

```text
Query : 2------5
Node  : 2------5
```

Return the stored value.

---

### 3. Partial Overlap

```text
Query :    2------5
Node  : 0------3
```

Split into children.

---

## Build Process

1. If the segment contains one element → store it.
2. Otherwise:
   - Split into left and right halves.
   - Build both children.
   - Combine their values.

---

## Query Process

1. No overlap → return identity value (e.g., `0` for sum).
2. Complete overlap → return node value.
3. Partial overlap → query both children and combine results.

---

## Update Process

1. Go to the leaf containing the index.
2. Update its value.
3. Update all ancestors while returning.

---

## Changing the Operation

Only the combine operation changes.

### Sum

```python
left + right
```
### Minimum

```python
min(left, right)
```
### Maximum

```python
max(left, right)
```
### XOR

```python
left ^ right
```
### GCD

```python
math.gcd(left, right)
```
### LCM

```python
(left * right) // math.gcd(left, right)
```
### Counting Frequency

```python
left + right
```

## Recursive Range Sum Segment Tree

````python
class SegmentTree:
    def __init__(self, arr):
        self.n = len(arr)
        self.arr = arr[:]
        self.tree = [0]*(4*self.n)
        self.build(1, 0, self.n - 1)

    # Build the tree
    def build(self, node, start, end):
        if start == end:
            self.tree[node] = self.arr[start]
            return

        mid = (start + end) // 2

        self.build(2 * node, start, mid)
        self.build(2 * node + 1, mid + 1, end)

        self.tree[node] = (
            self.tree[2 * node]
            + self.tree[2 * node + 1]
        )

    # Range Sum Query
    def query(self, node, start, end, left, right):
        # No overlap
        if right < start or end < left:
            return 0

        # Complete overlap
        if left <= start and end <= right:
            return self.tree[node]

        # Partial overlap
        mid = (start + end) // 2

        return (
            self.query(2 * node, start, mid, left, right)
            + self.query(2 * node + 1, mid + 1, end, left, right)
        )

    # Point Update
    def update(self, node, start, end, idx, value):
        if start == end:
            self.arr[idx] = value
            self.tree[node] = value
            return

        mid = (start + end) // 2

        if idx <= mid:
            self.update(2 * node, start, mid, idx, value)
        else:
            self.update(2 * node + 1, mid + 1, end, idx, value)

        self.tree[node] = (
            self.tree[2 * node]
            + self.tree[2 * node + 1]
        )


# ---------------- Example ----------------

arr = [5, 2, 7, 1]

st = SegmentTree(arr)

print(st.query(1, 0, 3, 1, 3))   # 10

st.update(1, 0, 3, 2, 10)

print(st.query(1, 0, 3, 1, 3))   # 13
````

## Iterative Segment Tree

````python
class SegmentTree:
    def __init__(self, arr):
        self.n = len(arr)
        self.tree = [0] * (2 * self.n)

        # Store array in leaves
        for i in range(self.n):
            self.tree[self.n + i] = arr[i]

        # Build parents
        for i in range(self.n - 1, 0, -1):
            self.tree[i] = self.tree[2 * i] + self.tree[2 * i + 1]

    def update(self, pos, value):
        pos += self.n
        self.tree[pos] = value

        while pos > 1:
            pos //= 2
            self.tree[pos] = self.tree[2 * pos] + self.tree[2 * pos + 1]

    def query(self, left, right):
        left += self.n
        right += self.n

        ans = 0

        while left <= right:
            if left % 2 == 1:
                ans += self.tree[left]
                left += 1

            if right % 2 == 0:
                ans += self.tree[right]
                right -= 1

            left //= 2
            right //= 2

        return ans


# Example
arr = [5, 2, 7, 1]
st = SegmentTree(arr)

print(st.query(1, 3))   # 10

st.update(2, 10)

print(st.query(1, 3))   # 13
````

## Recursive Segment Tree(No Class)

````python
def build(node, start, end):
    if start == end:
        tree[node] = arr[start]
        return

    mid = (start + end)//2
    build(2*node, start, mid)
    build(2*node + 1, mid + 1, end)
    tree[node] = tree[2*node] + tree[2*node + 1]

def query(node, start, end, left, right):
    if right < start or end < left:     # No overlap
        return 0
    if left <= start and end <= right:    # Complete overlap
        return tree[node]
    mid = (start + end)//2          # Partial overlap
    return (query(2*node, start, mid, left, right) + query(2*node + 1, mid + 1, end, left, right))

def update(node, start, end, idx, value):
    if start == end:
        arr[idx] = value
        tree[node] = value
        return
    mid = (start + end)//2
    if idx <= mid:
        update(2*node, start, mid, idx, value)
    else:
        update(2*node + 1, mid + 1, end, idx, value)
    tree[node] = tree[2*node] + tree[2*node + 1]

# ---------------- Example ----------------

arr = [5, 2, 7, 1]
n = len(arr)
tree = [0]*(4*n)
build(1, 0, n-1)
print(query(1, 0, n-1, 1, 3))  # 10
update(1, 0, n-1, 2, 10)
print(query(1, 0, n-1, 1, 3))  # 13
````

## Iterative Segment Tree(No Class)  Use this

````python
arr = [5, 2, 7, 1]
n = len(arr)
tree = [0]*(2*n)
for i in range(n): tree[n+i] = arr[i]    # Build tree
for i in range(n-1, 0, -1): tree[i] = tree[2*i] + tree[2*i + 1]    # Build parents
print("Tree:", tree)

def query(left, right):
    left += n; right += n; ans = 0
    while left <= right:
        if left % 2 == 1: ans += tree[left]; left += 1        # left is a right child
        if right % 2 == 0: ans += tree[right]; right -= 1     # right is a left child
        left //= 2; right //= 2
    return ans
print(query(1, 3))      # 10

def update(pos, value):
    pos += n; tree[pos] = value
    while pos > 1:
        pos //= 2
        tree[pos] = tree[2*pos] + tree[2*pos + 1]

update(2, 10)
print(query(1, 3))      # 13
print("Updated Tree:", tree)
````
