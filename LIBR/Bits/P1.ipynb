### **Right Shift Operator**, 

#### For a given n in it's binary representation removes the rightmost bit, which is simply n//2

```python
n = 42
print(bin(n)[2:])         # 101010

print(n >> 1)             # 21
print(bin(n >> 1)[2:])    # 10101

print(n >> 2)             # 10
print(bin(n >> 2)[2:])    # 1010

print(n >> 3)             # 5
print(bin(n >> 3)[2:])    # 101
```

### **Left Shift Operator** 
#### For a given integer n in it's binary representation adds bit 0 in the right, which ic simply multiplying n by 2

```python
n = 42                  
print(bin(n)[2:])       # 101010

print(n << 1)           # 84
print(bin(n << 1)[2:])  # 1010100

print(n << 2)           # 168
print(bin(n << 2)[2:])  # 10101000
```

### **Check, Set, Clear, Toggle**

```python
n = 43
i = 1

print("original :", bin(n)[2:], i)
print("check    :", (n >> i) & 1)             # i = 1 checks second bit from the right 0 or 1
print("set      :", bin(n | (1 << i))[2:])    # i = 1 changes second bit from the right from 0 to 1, if 1 then unchanged
print("clear    :", bin(n & ~(1 << i))[2:])   # i = 1 clears, sets to 0 the second bit from the right
print("toggle   :", bin(n ^ (1 << i))[2:])    # i = 1 flips second bit from the right from 0 to 1 or 1 to 0
print("\n")

i = 2
print("original :", bin(n)[2:], i)
print("check    :", (n >> i) & 1)             # i = 2 checks third bit from the right 0 or 1
print("set      :", bin(n | (1 << i))[2:])    # i = 2 changes third bit from the right from 0 to 1, if 1 then unchanged
print("clear    :", bin(n & ~(1 << i))[2:])   # i = 2 clears, sets to 0 the third bit from the right
print("toggle   :", bin(n ^ (1 << i))[2:])    # i = 2 flips third bit from the right from 0 to 1 or 1 to 0
```

### To turn the rightmost 1 to 0

### Starts from the right, if finds $10$ then changes to $01$

```python
n = 42; lis = [bin(n)[2:]]
for i in range(1, n.bit_length()):
    if (n >> i) & 1 and ((n >> (i - 1)) & 1) == 0:
        n ^= (1 << i) | (1 << (i - 1))
        lis.append(bin(n)[2:])
print(*lis)
```

### Starts from the right, if finds 01 then changes to 10

```python
n = 42; lis = [bin(n)[2:]]
for i in range(1, n.bit_length()):
    if ((n >> i) & 1) == 0 and ((n >> (i - 1)) & 1) == 1:
        n ^= (1 << i) | (1 << (i - 1))
        lis.append(bin(n)[2:])
print(*lis)
```

### For n given, changes only one 1 to 0 and all other bits remain same in the binary representation of n, starts from right to left

```python
n = 23; lis = [n]
for i in range(n.bit_length()):
    if (n >> i) & 1:
        lis.append(n ^ (1 << i))
print(*lis)
print(*[bin(i)[2:] for i in lis])
```

```python
# ===========================================================
# Problem 1
# max(min(j ^ i) for 0 <= i < j) for all 0 <= j <= ct
# ===========================================================

# Observation:
# The answer is the largest power of 2 <= ct.

def solve1(ct):
    return 1 << (ct.bit_length() - 1)


# Examples
# ct = 5  -> 4
# ct = 7  -> 4
# ct = 8  -> 8
# ct = 13 -> 8


# ===========================================================
# Problem 2
# max(min(j ^ i) for 0 <= i <= ct, i != j) for all 0 <= j <= ct
# ===========================================================

# Observation:
# The answer is the least significant set bit (lowest power of 2)
# present in ct.

def solve2(ct):
    return ct & -ct


# Examples
# ct = 5  -> 1
# ct = 6  -> 2
# ct = 8  -> 8
# ct = 12 -> 4


# ===========================================================
# Problem 3
# Given an array ct[]
#
# max(min(ct[j] ^ ct[i]) for 0 <= i < j)
# ===========================================================

# O(30 * n) Binary Trie

def solve3(ct):
    if len(ct) < 2:
        return 0

    B = 29                     # values <= 1e9

    left = [-1]
    right = [-1]

    def insert(x):
        node = 0
        for b in range(B, -1, -1):
            if (x >> b) & 1:
                if right[node] == -1:
                    right[node] = len(left)
                    left.append(-1)
                    right.append(-1)
                node = right[node]
            else:
                if left[node] == -1:
                    left[node] = len(left)
                    left.append(-1)
                    right.append(-1)
                node = left[node]

    def min_xor(x):
        node = 0
        ans = 0

        for b in range(B, -1, -1):
            bit = (x >> b) & 1

            # Prefer the same bit to minimize XOR.
            if bit:
                if right[node] != -1:
                    node = right[node]
                else:
                    ans |= 1 << b
                    node = left[node]
            else:
                if left[node] != -1:
                    node = left[node]
                else:
                    ans |= 1 << b
                    node = right[node]

        return ans

    insert(ct[0])

    ans = 0
    for x in ct[1:]:
        ans = max(ans, min_xor(x))
        insert(x)

    return ans


# ===========================================================
# Useful bit facts used above
# ===========================================================

# Largest power of 2 <= x
largest_power = 1 << (x.bit_length() - 1)

# Lowest set bit of x
lowest_set_bit = x & -x

# Highest bit index of x
highest_bit = x.bit_length() - 1
```

## Problem on XOR

```python
# https://codeforces.com/contest/2247/problem/D1
import sys
f = open("input1.txt", "rt")
sys.stdin = f

t = sys.stdin
for _ in range(int(t.readline())):
    n, q = map(int, t.readline().split()); a = list(map(int, t.readline().split()))
    if a == sorted(a): print(0); continue
    x = 0
    while True:
        block_size = 1 << (x+1); mx = -1; good = True
        for i in range(0, n, block_size):
            cmx = -1
            for j in range(i, min(i+block_size, n)):
                good = good and (mx <= a[j])
                cmx = max(cmx, a[j])
            mx = max(mx, cmx)
        if good: print(1 << x); break
        x += 1
```
