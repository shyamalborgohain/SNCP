## Levenshtein (Edit) Distance

- **Use:** Minimum edits required to convert one string into another.
- **Operations:** Insert, Delete, Replace
- **Time Complexity:** O(nm)
- **Space Complexity:** O(nm) (O(m) with rolling array)

```python
def levenshtein(x, y):
    n, m = len(x), len(y)
    dp = [[0]*(m+1) for _ in range(n+1)]
    for i in range(n+1): dp[i][0] = i
    for j in range(m+1): dp[0][j] = j
    for i in range(1, n+1):
        for j in range(1, m+1):
            dp[i][j] = min(
                dp[i-1][j] + 1,                    # delete
                dp[i][j-1] + 1,                    # insert
                dp[i-1][j-1] + (x[i-1] != y[j-1])  # replace
            )
    return dp[n][m]

A = "kitten"    # changes A to B
B = "sitting"
print(levenshtein(A, B))   # 3
```

## Longest Common Subsequence Algorithm(LCS)

- **Time Complexity:** O(nm)
- **Space Complexity:** O(nm)

### For Longest Common Palindromic Subsequence(LCPS) of a string, Perform LCS(s, s[::-1]) 

### Given two strings s1 and s2, to find the length of the smallest string which has both s1 and s2 as its sub-sequences simply return len(s1)+len(s2)-len(LCS(s1, s2))

```python
def lcs(a, b):
    n = len(a); m = len(b)
    dp = [[0]*(m+1) for _ in range(n+1)]                          # dp[i][j] = LCS of a[:i], b[:j]
    for i in range(1, n + 1):                                     # fill the DP table        
        for j in range(1, m+1):
            if a[i-1] == b[j-1]: dp[i][j] = dp[i-1][j-1] + 1      # take match
            else: dp[i][j] = max(dp[i-1][j], dp[i][j-1])          # skip one
    
    i = n; j = m; lcs_seq = []                                             # backtracking to recover one LCS from the DP table
    while i > 0 and j > 0:
        if a[i-1] == b[j-1]: lcs_seq.append(a[i-1]); i -= 1; j -= 1        # Matching element belongs to the LCS
        elif dp[i - 1][j] >= dp[i][j - 1]: i -= 1                          # Move in the direction of the larger DP value
        else: j -= 1
    
    lcs_seq.reverse()
    return dp[n][m], lcs_seq

A = [0, 5, 2, 4, 8, 4]
B = [0, 5, 4, 8, 4]

length, sequence = lcs(A, B)
print("LCS Length :", length)
print("LCS        :", sequence)
```

### Given two strings str1 and str2, return the shortest string that has both str1 and str2 as subsequences

```python
class Solution:
    def shortestCommonSupersequence(self, str1: str, str2: str) -> str:
        n, m = len(str1), len(str2); dp = [[0]*(m+1) for _ in range(n+1)]
        for i in range(1, n+1):
            for j in range(1, m+1):
                if str1[i-1] == str2[j-1]: dp[i][j] = dp[i-1][j-1] + 1
                else: dp[i][j] = max(dp[i-1][j], dp[i][j-1])
        i = n; j = m; seq = ''
        while i > 0 and j > 0:
            if str1[i-1] == str2[j-1]: seq += str1[i-1]; i -= 1; j-= 1
            elif dp[i-1][j] >= dp[i][j-1]: seq += str1[i-1]; i -= 1
            else: seq += str2[j-1]; j -= 1
        while i > 0: seq += str1[i-1]; i -= 1
        while j > 0: seq += str2[j-1]; j -= 1
        return seq[::-1]

str1 = "abac"
str2 = "cab"
print(Solution().shortestCommonSupersequence(str1, str2))
```

### All Possible Subsequences

```python
class Solution:
    def numDistinct(self, s: str, t: str) -> int:
        m = len(t); dp = [0]*(m+1); dp[0] = 1
        for ch in s:
            for j in range(m, 0, -1):
                if ch == t[j-1]: dp[j] += dp[j-1]
        return dp[m]
```

```python
from functools import lru_cache
def all_lcs(a, b):
    n, m = len(a), len(b)
    dp = [[0]*(m+1) for _ in range(n+1)]   # LCS length

    for i in range(1, n+1):
        for j in range(1, m+1):
            if a[i-1] == b[j-1]: dp[i][j] = dp[i-1][j-1] + 1
            else: dp[i][j] = max(dp[i-1][j], dp[i][j-1])

    @lru_cache(None)
    def backtrack(i, j):
        if i == 0 or j == 0: return {()}
        if a[i-1] == b[j-1]: return {seq + (a[i-1],) for seq in backtrack(i-1, j-1)}
        ans = set()
        if dp[i-1][j] == dp[i][j]: ans |= backtrack(i-1, j)
        if dp[i][j-1] == dp[i][j]: ans |= backtrack(i, j-1)
        return ans
    sequences = [list(seq) for seq in backtrack(n, m)]
    return dp[n][m], sequences

A = list("ABCBDAB")
B = list("BDCABA")

length, lcs_list = all_lcs(A, B)
print("Length:", length)
for seq in sorted(lcs_list): print("".join(seq))
```

## Longest Increasing Subsequence

- **Time Complexity:** DP: $O(n^2)$, Binary Search: $O(nlogn)$
- **Space Complexity:** $O(n)$

### DP Algorithm

```python
def lisa(a):
    n = len(a)
    dp = [1]*n          # LIS ending at i
    parent = [-1]*n     # Previous index in LIS

    for i in range(n):
        for j in range(i):
            if a[j] < a[i] and dp[j] + 1 > dp[i]: dp[i] = dp[j] + 1; parent[i] = j
    length = max(dp)
    idx = dp.index(length)
    seq = []
    while idx != -1: seq.append(a[idx]); idx = parent[idx]
    seq.reverse()
    return length, seq

A = [10, 9, 2, 5, 3, 7, 101, 18]

length, seq = lisa(A)
print(length)   # 4
print(seq)      # [2, 5, 7, 101]
```

### LIS Binary Search Algorithm

#### Use bisect_right if you want Longest Non-Decreasing Subsequence 

```python
from bisect import bisect_left

def lisb(a):
    tail = []             # Smallest tail of LIS of each length
    tail_idx = []         # Index of each tail
    parent = [-1]*len(a)

    for i, x in enumerate(a):
        pos = bisect_right(tail, x)
        if pos == len(tail): tail.append(x); tail_idx.append(i)
        else: tail[pos] = x; tail_idx[pos] = i
        if pos > 0: parent[i] = tail_idx[pos-1]
        
    seq = []; idx = tail_idx[-1]
    while idx != -1: seq.append(a[idx]); idx = parent[idx]
    seq.reverse()
    return len(tail), seq

A = [10, 9, 2, 6, 3, 7, 101, 18]
B = [2, 8, 4, 9, 5, 10, 3, 11] 

length, seq = lisb(B)
print(length)   # 4
print(seq)      # [2, 3, 7, 18]
```

## Lowest Common Substring

- Characters must be contiguous
- **Time Complexity:** O(nm)
- **Space Complexity:** O(nm)

```python
def longest_common_substring(a, b):
    n, m = len(a), len(b)
    dp = [[0]*(m+1) for _ in range(n+1)]
    best = 0; end = 0

    for i in range(1, n+1):
        for j in range(1, m+1):
            if a[i-1] == b[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
                if dp[i][j] > best: best = dp[i][j]; end = i
                
    return a[end-best : end]

a = "abcdef"
b = "zabcf"
print(longest_common_substring(a, b))
```

## Longest Contigous Substring whose characters can be rearranged to form a palindrome

```python
class Solution:
    def LSFP(self, s):
        first = {0: -1}; mask = 0; ans = 0

        for i, c in enumerate(s):
            mask ^= 1 << (ord(c) - ord('a'))

            # Case 1: all frequencies even
            if mask in first: ans = max(ans, i - first[mask])

            # Case 2: exactly one odd frequency
            for b in range(26):
                other = mask ^ (1 << b)

                if other in first: ans = max(ans, i - first[other])

            # Store earliest occurrence only
            if mask not in first: first[mask] = i

        return ans
```

```python
tests = ["aabbccx", "aabb", "abc", "a", "aa", "abca", "aabbccd", "abcba", "aabbc", "abcdef",]

for s in tests:
    result = Solution().LSFP(s)
    print(s, result)
```

## Manacher's algorithm for Longest Palindromic Substring

```python
class Solution:
    def longestPalindrome(self, s):
        if not s:
            return ""

        # Add separators so odd/even palindromes are handled uniformly
        t = "^#" + "#".join(s) + "#$"

        n = len(t)
        p = [0] * n

        center = 0
        right = 0

        for i in range(1, n - 1):

            mirror = 2 * center - i

            # Use previously calculated palindrome
            if i < right:
                p[i] = min(right - i, p[mirror])

            # Expand around i
            while t[i + 1 + p[i]] == t[i - 1 - p[i]]:
                p[i] += 1

            # Update center and right boundary
            if i + p[i] > right:
                center = i
                right = i + p[i]

        # Find largest palindrome
        max_len = max(p)
        center_idx = p.index(max_len)

        start = (center_idx - max_len) // 2

        return s[start:start + max_len]
```

```python
s = "babad"

print(Solution().longestPalindrome(s))
```
