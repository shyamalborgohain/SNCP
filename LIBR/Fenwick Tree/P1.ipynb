## Fenwick Tree (Binary Indexed Tree)

- **Point Update** — `O(log n)`
- **Prefix Query** — `O(log n)`
- Space: **O(n)**

### Key Idea: 
- Uses **1-based indexing**.
- Lowest Set Bit (LSB): i & -i

### Traversal

- **Update** (move up): i += i & -i
- **Query** (move down): i -= i & -i

### Range Query

- Range(L, R) = Prefix(R) - Prefix(L-1)

### Supports

- Prefix Sum
- Range Sum
- Prefix Maximum
- Frequency Counting
- Coordinate Compression
- Inversion Count
- DP Optimizations

### Complexity

| Operation | Update | Query | Space |
|-----------|------|----|---|
| Time | O(log n) | O(log n) | O(n) |

### Remember

- **Sum BIT:** use `+=`
- **Max BIT:** use `max()`
- Always **1-indexed**
- `i & -i` gives the Lowest Set Bit (LSB)


```python
n = 5
arr = [1, 2, 3, 4, 5]
bit = [0]*(n+1)

def update(i, val):
    while i <= n:
        bit[i] += val
        i += i & -i

def query(i):
    s = 0
    while i:
        s += bit[i]
        i -= i & -i
    return s

def range_sum(l, r):
    return query(r) - query(l - 1)

# Build Fenwick Tree
for i, x in enumerate(arr, 1):
    update(i, x)

print(query(3))          # 6
print(range_sum(2, 5))   # 14
update(3, 2)             # arr[3] += 2
print(query(3))          # 8
```

## Example Problem

```python
# https://codeforces.com/problemset/problem/2244/G
def update(pos, val):
    while pos <= n: bit[pos] = max(bit[pos], val); pos += pos & -pos

def query(pos):
    ans = 0
    while pos: ans = max(ans, bit[pos]); pos -= pos & -pos
    return ans

import sys
f = open("input1.txt", "rt")
sys.stdin = f

t = sys.stdin
for _ in range(int(t.readline())):
    n = int(t.readline()); a = [0] + list(map(int, t.readline().split()))
    bit = [0]*(n+1); events = [[] for _ in range(n+1)]; ans = 0
    for i in range(1, n+1):
        for pos, val in events[i]: update(pos, val)
        l = i - a[i] - 1; mx = query(min(n, l)) if l > 0 else 0
        dp = a[i] + mx; ans = max(ans, dp); act = i + a[i] + 1
        if act <= n: events[act].append((i, dp))
    print(ans)
```
