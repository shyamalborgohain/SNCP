## [Search in Rotated Sorted Array II (duplicates are present) (GeeksforGeeks)](https://www.geeksforgeeks.org/problems/search-in-rotated-array-2/1)

- #### Worst Case is similar to using iteration

```python
from bisect import bisect_left
class Solution:
    def search(self, arr, key):
        n = len(arr); lo = 0; hi = n-1
        while lo < hi:
            mid = lo + (hi-lo)//2
            if arr[mid] > arr[hi]: lo = mid+1
            elif arr[mid] < arr[hi]: hi = mid
            else: hi -= 1
        pivot = lo
        idx1 = bisect_left(arr, key, 0, pivot)
        if idx1 < n and arr[idx1] == key: return True
        idx2 = bisect_left(arr, key, pivot, n)
        if idx2 < n and arr[idx2] == key: return True
        return False
```

```python
arr = [3, 3, 3, 1, 2, 3]; key = 3
print(Solution().search(arr, key))
```

## [540. Single Element in a Sorted Array (LeetCode)](https://leetcode.com/problems/single-element-in-a-sorted-array/description/) 

```python
class Solution:
    def singleNonDuplicate(self, arr):
        lo = 0; hi = len(arr) - 1
        while lo < hi: 
            mid = lo + (hi-lo)//2
            if mid%2 == 1: mid -= 1
            if arr[mid] == arr[mid+1]: lo = mid+2
            else: hi = mid
        return arr[lo]
```

```python
nums = [1,1,2,3,3,4,4,8,8]
print(Solution().singleNonDuplicate(nums))
```
