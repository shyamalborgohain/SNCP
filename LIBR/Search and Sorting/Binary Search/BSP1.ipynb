## Binary Search

```python
# Recursive
def binarySearch1(arr, low, high, target):
    if low > high: return -1
    mid = low + (high - low)//2
    if arr[mid] == target: return mid
    elif arr[mid] < target: return binarySearch1(arr, mid + 1, high, target)
    else: return binarySearch1(arr, low, mid - 1, target)
```

```python
arr = [10, 20, 30, 40, 50, 60, 70]
print(binarySearch1(arr, 0, len(arr)-1, 50))
```

```python
# Iterative
def binarySearch2(arr, target):
    low = 0; high = len(arr) - 1
    while low <= high: 
        mid = low + (high - low)//2
        if arr[mid] == target: return mid
        elif arr[mid] < target: low = mid + 1
        else: high = mid - 1
    return -1
```

```python
arr = [10, 20, 30, 40, 50, 60, 70]
print(binarySearch2(arr, 50))
```

## Binary Search Using Python's bisect

```python
import bisect
arr = [10, 20, 30, 40, 50, 60, 70]
print(bisect.bisect_left(arr, 50))
```
