## [Longest Substring with Distinct Characters](https://www.geeksforgeeks.org/problems/longest-distinct-characters-in-string5848/1)

````python
class Solution:
    def longestUniqueSubstr(self, s):
        seen = set(); left = 0; mx = 0
        for right in range(len(s)):
            while s[right] in seen: seen.remove(s[left]); left += 1
            seen.add(s[right]); mx = max(mx, right - left + 1)
        return mx
````

````python
s = "geiifhgf"
print(Solution().longestUniqueSubstr(s))
````

## [Maximum you can obtain from Cards](https://www.geeksforgeeks.org/problems/maximum-point-you-can-obtain-from-cards/1)

#### Use **Auxillary Space** $O(1)$, if ```sum(len[:k])``` is used, then auxillary space taken is $O(k)$, because it creates a new list containing the first k elements.

````python
class Solution:
    def maxScore(self, lis, k):
        n = len(lis); left = 0; right = n; val = 0
        for i in range(k): val += lis[i]
        mx = val
        if k == n: return val
        for i in range(k): val += lis[-i-1] - lis[k-i-1]; mx = max(mx, val)
        return mx
````

````python
lis = [1, 2, 3, 4, 5, 6, 1]; k = 3
print(Solution().maxScore(lis, k))
````

## [Maximum Consecutive 1's with k flips](https://www.geeksforgeeks.org/problems/maximize-number-of-1s0905/1)

````python
class Solution:
    def maxOnes(self, arr, k):
        n = len(arr); left = 0; mx = 0
        for right in range(n):
            if arr[right] == 0: k -= 1
            while k < 0: 
                if arr[left] == 0: k += 1
                left += 1
            mx = max(mx, right - left + 1)
        return mx
````

````python
arr = [1, 0, 0, 1, 0, 1]; k = 0
print(Solution().maxOnes(arr, k))
````

## [Fruits and Baskets](https://www.naukri.com/code360/problems/fruits-and-baskets_985356?utm_source=chatgpt.com&leftPanelTabValue=SUBMISSION)

````python
class Solution:
    def findMaxFruits(self, arr, n):
        left = mid = mx = 0
        for right in range(n):
            if arr[right] != arr[mid]:
                if arr[right] != arr[left]: left = mid
                mid = right
            mx = max(mx, right - left + 1)
        return mx
````

````python
arr = [1, 1, 2, 2, 2, 1, 3, 4]; n = 8
print(Solution().findMaxFruits(arr, n))
````
