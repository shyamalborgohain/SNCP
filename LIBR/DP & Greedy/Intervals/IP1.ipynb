## Interval Scheduling Problem (Greedy)

- **Problem:** Given a set of activities/intervals, each with a start time and finish time, select the maximum number of non-overlapping intervals.

- **Time Complexity:** $O(n\log{n})$ and **Space Complexity:** $O(n)$

```python
def intervalScheduling(intervals):
    intervals.sort(key=lambda x: x[1])
    selected = []; last_finish = float('-inf')
    for start, finish in intervals:
        if start >= last_finish: selected.append((start, finish)); last_finish = finish
    return len(selected), selected
```

```python
intervals = [(1, 3), (2, 5), (4, 7), (6, 8), (8, 10)]
print(*intervalScheduling(intervals))
```

## Interval Covering Problem (Greedy)

- **Problem:** Given a set of intervals and a target interval $[L, R]$, select the minimum number of given intervals whose union completely covers $[L, R]$.


- **Time Complexity:** $O(n \log n)$ and **Space Complexity:** $O(n)$

```python
def intervalCover(intervals, L, R):
    intervals.sort(key = lambda x: x[0]); selected = []; i = 0; cur = L; n = len(intervals)
    while cur < R:
        farthest = cur; best = None
        while i < n and intervals[i][0] <= cur:
            if intervals[i][1] > farthest: farthest = intervals[i][1]; best = intervals[i]
            i += 1
        if farthest == cur: return None
        selected.append(best); cur = farthest
    return len(selected), selected
```

```python
intervals = [(0, 2), (1, 5), (3, 7), (5, 9), (7, 10)]
print(intervalCover(intervals, 0, 10))
```

## Scheduling to Minimize Lateness (Greedy)

- **Problem:** Given jobs with: processing time $(p)$ and deadline $(d)$.Schedule all jobs on a single machine such that maximum lateness is minimized.

- **Time Complexity:** $O(n \log n)$ and **Space Complexity:** $O(1)$

```python
def minimizeLateness(jobs):
    jobs.sort(key = lambda x: x[2]); time = 0; max_lateness = 0
    for job, p, d in jobs: time += p; max_lateness = max(max_lateness, time - d)
    return max_lateness
```

```python
jobs = [("A", 3, 4), ("B", 2, 6), ("C", 1, 5)]
print(minimizeLateness(jobs))
```

## Weighted Interval Scheduling 

- **Problem:** Given intervals/jobs, each having: start time, finish time, weight/profit. Select non-overlapping intervals such that the total weight is maximum.

- **Time Complexity:** $O(n^2)$ and **Space Complexity:** $O(n)$

```python
def weightedIntervalScheduling(intervals):
    intervals.sort(key = lambda x: x[1]); n = len(intervals); dp = [0]*(n+1)
    for i in range(1, n+1):
        start, finish, weight = intervals[i-1]; j = i-2
        while j >= 0 and intervals[j][1] > start: j -= 1
        dp[i] = max(dp[i-1], dp[j+1] + weight)
    return dp[-1]
```

```python
jobs = [(1, 3, 5), (2, 5, 6), (4, 6, 5), (6, 8, 4), (7, 9, 7)]
print(weightedIntervalScheduling(jobs))
```

## Max Intersection Point

- **Problem:** To find the point that lies in the maximum number of intervals.

- **Time Complexity:** $O(n \log n)$ and **Space Complexity:** $O(n)$

```python
def maxIntersectionPoint(intervals):
    points = []
    for start, end in intervals: points.append((start, 1)); points.append((end, -1))
    points.sort(key=lambda x: x[0]); count = 0; max_count = 0; point = None
    for i in range(len(points)):
        count += points[i][1]
        if count > max_count: max_count = count; point = points[i][0]
    return point, max_count
```

```python
intervals = [(1, 5), (2, 6), (3, 4), (4, 7)]
print(*maxIntersectionPoint(intervals))
```

## Interval Union 

- **Problem:** Given a set of intervals, find their union by merging all overlapping intervals.

- **Time Complexity:** $O(n \log n)$ and **Space Complexity:** $O(n)$

```python
def intervalUnion(intervals):
    intervals.sort(key=lambda x: x[0]); result = []
    for start, end in intervals:
        if not result or start > result[-1][1]: result.append([start, end])
        else: result[-1][1] = max(result[-1][1], end)
    return result
```

```python
intervals = [(1, 3), (2, 6), (8, 10), (9, 12)]
print(*intervalUnion(intervals))
```

## Interval Point Cover

```python
def pointCover(intervals):
    intervals.sort(key=lambda x: x[1]); points = []
    for start, end in intervals:
        if not points or points[-1] < start: points.append(end)
    return points
```

```python
intervals = [(1, 3), (2, 5), (4, 6)]
print(pointCover(intervals))
```
