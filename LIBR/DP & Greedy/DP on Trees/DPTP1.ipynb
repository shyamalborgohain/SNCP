## Maximum Independent Set in a Tree

An **Independent Set** is a set of vertices such that no two selected vertices are directly connected by an edge.

- **Problem:** Given a tree with n vertices, find the maximum number of vertices an **Independent Set** can have.

- **Time Complexity:** $O(n)$ and **Space Complexity:** $O(n)$

#### **Remark:** For Maximum Weighted Independent Set replace $1$ by the weight, $w_i$.

#### **Idea:** Use recursion or iteration(better)

For each node $u$, maintain two values:

- $dp[u][0]$ = maximum independent set in the subtree of $u$ when $u$ is **not selected**.
- $dp[u][1]$ = maximum independent set in the subtree of \(u\) when $u$ **is selected**.

$$
dp[u][1] = 1+\sum_{v\in children(u)} dp[v][0]
$$
$$
dp[u][0] = \sum_{v\in children(u)} \max\left(dp[v][0],dp[v][1]\right)
$$

```python
def maxIndependentSet1(tree, u, parent):
    dp0 = 0; dp1 = 1
    for v in tree[u]:
        if v != parent:
            a, b = maxIndependentSet1(tree, v, u); dp0 += max(a, b); dp1 += a
    return dp0, dp1
```

```python
tree = [[1, 2], [0, 3, 4], [0], [1], [1]]
a, b = maxIndependentSet1(tree, 0, -1)
print(max(a, b))
```

```python
def maxIndependentSet2(tree, root):
    n = len(tree); parent = [-1]*n; order = []; stack = [root]
    while stack:
        u = stack.pop(); order.append(u)
        for v in tree[u]:
            if v != parent[u]: parent[v] = u; stack.append(v)
    dp0 = [0]*n; dp1 = [1]*n
    for u in reversed(order):
        for v in tree[u]:
            if parent[v] == u: dp0[u] += max(dp0[v], dp1[v]); dp1[u] += dp0[v]
    return max(dp0[root], dp1[root])
```

```python
tree = [[1, 2], [0, 3, 4], [0], [1], [1]]
print(maxIndependentSet2(tree, 0))
```

## Minimum Dominating Set in a Tree

A **dominating set** is a set of vertices such that every vertex is either selected or adjacent to a selected vertex.

**Problem:** Given a tree, find the **minimum number of vertices** needed to form a dominating set.

- **Time Complexity:** $O(n)$ and **Space Complexity:** $O(n)$

#### **Remark:** For Weighted Minimum Dominating Set replace $1$ by the weight, $w_i$.

For each node $u$, use three states:

- $dp[u][0]$: $u$ is **selected**.
- $dp[u][1]$: $u$ is **not selected**, but is **dominated by a child**.
- $dp[u][2]$: $u$ is **not selected** and needs its **parent to dominate it**.

$$
dp[u][0] = 1+\sum_{v\in children(u)} \min(dp[v][0],dp[v][1],dp[v][2])
$$
$$
\qquad dp[u][2] = \sum_{v\in children(u)} \min(dp[v][0],dp[v][1])
$$
$$
\qquad dp[u][1] = dp[u][2] + extra = \sum_{v\in children(u)} \min(dp[v][0],dp[v][1]) + min_{v \in children(u)}(extra, dp0[v] - min(dp0[v], dp1[v])))
$$

For the root, state $2$ is not allowed because it has no parent.

```python
def minDominatingSet(tree, root):
    n = len(tree); parent = [-1]*n; order = []; stack = [root]; dp0 = [1]*n; dp1 = [0]*n; dp2 = [0]*n
    while stack:
        u = stack.pop(); order.append(u)
        for v in tree[u]:
            if v != parent[u]: parent[v] = u; stack.append(v)
    for u in reversed(order):
        extra = 10**31
        for v in tree[u]:
            if parent[v] == u: 
                dp0[u] += min(dp0[v], dp1[v], dp2[v]); dp2[u] += min(dp0[v], dp1[v]); extra = min(extra, dp0[v] - min(dp0[v], dp1[v]))
        dp1[u] = dp2[u] + extra
    return min(dp0[root], dp1[root])
```

```python
tree = [[1, 2], [0, 3, 4], [0], [1], [1]]
print(minDominatingSet(tree, 0))
```

## Tree Vertex Cover

- **Problem:** Find the minimum number of vertices such that every edge has at least one selected endpoint.

- **Time Complexity:** $O(n)$ and **Space Complexity:** $O(n)$

```python
def vertexCover(tree, root):
    n = len(tree); parent = [-1]*n; order = []; stack = [root]; dp0 = [1]*n; dp1 = [0]*n
    while stack:
        u = stack.pop(); order.append(u)
        for v in tree[u]:
            if v != parent[u]: parent[v] = u; stack.append(v)
    for u in reversed(order):
        for v in tree[u]:
            if parent[v] == u: dp0[u] += min(dp0[v], dp1[v]); dp1[u] += dp0[v]
    return min(dp0[root], dp1[root])
```

```python
tree = [[1, 2], [0, 3, 4], [0], [1], [1]]
print(vertexCover(tree, 0))
```
