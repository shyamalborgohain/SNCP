## [Search in Rotated Sorted Array II (duplicates are present) (GeeksforGeeks)](https://www.geeksforgeeks.org/problems/search-in-rotated-array-2/1)

- #### Worst Case is similar to using iteration

```python
from bisect import bisect_left
class Solution:
    def search(self, arr, key):
        n = len(arr); lo = 0; hi = n-1
        while lo < hi:
            mid = lo + (hi-lo)//2
            if arr[mid] > arr[hi]: lo = mid+1
            elif arr[mid] < arr[hi]: hi = mid
            else: hi -= 1
        pivot = lo
        idx1 = bisect_left(arr, key, 0, pivot)
        if idx1 < n and arr[idx1] == key: return True
        idx2 = bisect_left(arr, key, pivot, n)
        if idx2 < n and arr[idx2] == key: return True
        return False
```

```python
arr = [3, 3, 3, 1, 2, 3]; key = 3
print(Solution().search(arr, key))
```

## [540. Single Element in a Sorted Array (LeetCode)](https://leetcode.com/problems/single-element-in-a-sorted-array/description/) 

```python
class Solution:
    def singleNonDuplicate(self, arr):
        lo = 0; hi = len(arr) - 1
        while lo < hi: 
            mid = lo + (hi-lo)//2
            if mid%2 == 1: mid -= 1
            if arr[mid] == arr[mid+1]: lo = mid+2
            else: hi = mid
        return arr[lo]
```

```python
nums = [1,1,2,3,3,4,4,8,8]
print(Solution().singleNonDuplicate(nums))
```

## [162. Find Peak Element(LeetCode)](https://leetcode.com/problems/find-peak-element/)

```python
class Solution:
    def findPeakElement(self, nums: list[int]) -> int:
        lo = 0; hi = len(nums)-1
        while lo < hi:
            mid = lo + (hi-lo)//2
            if nums[mid] < nums[mid+1]: lo = mid+1
            else: hi=mid
        return lo
```

```python
nums = [1,2,1,3,4,5,6,7,8,9,10]
print(Solution().findPeakElement(nums))
```

## [Search in Row-Column Sorted](https://www.geeksforgeeks.org/problems/search-in-a-matrix17201720/1)

#### Time Complexity: **StairCase Method** has $O(n+m)$, while **Binary Search** one have $O(n \log m)$ or $O(m \log n)$

```python
class Solution:
    def matSearch1(self, mat, target):
        rows = len(mat); cols = len(mat[0]); row = 0; col = cols - 1
        while row < rows and col >= 0:
            x = mat[row][col]
            if x == target: return True
            elif x > target: col -= 1
            else: row += 1
        return False
```

```python
mat = [[3, 30, 38], [20, 52, 54], [35, 60, 69]]; x = 62
print(Solution().matSearch1(mat, x))
```

### Using Binary Search

```python
class Solution:
    def matSearch2(self, mat, target):
        rows = len(mat); cols = len(mat[0])
        for row in range(rows):
            lo = 0; hi = cols - 1
            while lo <= hi:
                mid = lo + (hi-lo)//2
                if mat[row][mid] == target: return True
                elif mat[row][mid] < target: lo = mid + 1
                else: hi = mid - 1
        return False
```

```python
mat = [[3, 30, 38], [20, 52, 54], [35, 60, 69]]; x = 62
print(Solution().matSearch2(mat, x))
```

## [Find a Peak Element II](https://leetcode.com/problems/find-a-peak-element-ii/description/)

```python
class Solution:
    def findPeakGrid(self, mat):
        rows = len(mat); cols = len(mat[0]); lo = 0; hi = cols - 1
        while lo <= hi:
            mid = lo + (hi-lo)//2; max_row = 0
            for r in range(1, rows):
                if mat[r][mid] > mat[max_row][mid]: max_row = r
            left = mat[max_row][mid-1] if mid > 0 else -1
            right = mat[max_row][mid+1] if mid < cols-1 else -1
            if mat[max_row][mid] > left and mat[max_row][mid] > right: return [max_row, mid]
            elif mat[max_row][mid] < right: lo = mid + 1
            else: hi = mid - 1
```

```python
mat = [[10,20,15],[21,30,14],[7,16,32]]
print(Solution().findPeakGrid(mat))
```

## [k $^{th}$ Missing Positive Number in a Sorted Array](https://www.geeksforgeeks.org/problems/kth-missing-positive-number-in-a-sorted-array/1)

```python
class Solution:
    def kthMissing(self, arr, k):
        lo = 0; hi = len(arr)-1
        while lo <= hi:
            mid = lo + (hi-lo)//2
            if arr[mid]-(mid+1) < k: lo = mid+1
            else: hi = mid-1
        return lo+k
```

```python
arr = [1, 2, 3]; k = 2
print(Solution().kthMissing(arr, k))
```

## [Aggressive Cows](https://www.geeksforgeeks.org/problems/aggressive-cows/1)

#### This same pattern appears in **Book Allocation**, **Painter's Partition**, **Split Array Largest Sum** and many other.

```python
class Solution:
    def aggressiveCows(self, arr, k):
        arr.sort(); lo = 1; hi = arr[-1] - arr[0]; ans = 0; n = len(arr)
        while lo <= hi:
            mid = lo + (hi-lo)//2
            if self.canAssign(arr, k, mid, n): ans = mid; lo = mid + 1
            else: hi = mid - 1
        return ans
    
    def canAssign(self, arr, k, dist, n):
        cows = 1; prev = arr[0]
        for i in range(1, n):
            if arr[i] - prev >= dist: cows += 1; prev = arr[i]
            if cows == k: return True
        return False
```

```python
arr = [10, 1, 2, 7, 5]; k = 3
print(Solution().aggressiveCows(arr, k))
```

## [Allocate Minimum Pages](https://www.geeksforgeeks.org/problems/allocate-minimum-number-of-pages0937/1)

```python
class Solution:
    def findPages(self, arr, k):
        n = len(arr)
        if k > n: return -1
        lo = max(arr); hi = sum(arr)
        while lo <= hi: 
            mid = lo + (hi-lo)//2
            if self.canAllocate(arr, k, mid): ans = mid; hi = mid - 1
            else: lo = mid + 1
        return ans
    
    def canAllocate(self, arr, k, max_pages):
        students = 1; pages = 0
        for book in arr:
            if pages + book <= max_pages: pages += book
            else:
                students += 1; pages = book
                if students > k: return False
        return True
```

```python
arr = [12, 34, 67, 90]; k = 2
print(Solution().findPages(arr, k))
```
