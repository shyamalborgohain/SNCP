## [Pre-order Traversal](https://www.geeksforgeeks.org/problems/preorder-traversal/1)

#### A preorder traversal first visits the node, then visits the left child (including its entire subtree), and finally visits the right child (including its entire subtree).

```python
class Node:
    def __init__(self,val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def preOrder(self, root):
        ans = []
        def dfs(node):
            if node is None: return
            ans.append(node.data)
            dfs(node.left)
            dfs(node.right)
        dfs(root)
        return ans

root = [6, 3, 2, None, 1, 2, None]
print(Solution().preOrder(root))
```

```python
# root = [6, 3, 2, None, 1, 2, None]
# Create the tree
root = Node(6)
root.left = Node(3)
root.right = Node(2)

root.left.left = None
root.left.right = Node(1)

root.right.left = Node(2)
root.right.right = None

print(Solution().preOrder(root))
```

```python
class Node:
    def __init__(self,val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def preOrder_iter(self, root):
        if root is None: return []
        ans = []; stack = [root]
        while stack:
            node = stack.pop(); ans.append(node.data)
            if node.right: stack.append(node.right)
            if node.left: stack.append(node.left)
        return ans
```

```python
# Create the tree
root = Node(6)
root.left = Node(3)
root.right = Node(2)

root.left.left = None
root.left.right = Node(1)

root.right.left = Node(2)
root.right.right = None

print(Solution().preOrder_iter(root))
```

## [In-order Traversal](https://www.geeksforgeeks.org/problems/inorder-traversal/1)

#### An inorder traversal first visits the left child (including its entire subtree), then visits the node, and finally visits the right child (including its entire subtree).

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def inOrder(self, root):
        ans = []
        def dfs(node):
            if node is None: return
            dfs(node.left)
            ans.append(node.data)
            dfs(node.right)
        dfs(root)
        return ans
```

```python
# root = [8, 1, 5, None, 7, 10, 6, None, 10, 6]
root = Node(8)
root.left = Node(1)
root.right = Node(5)

root.left.left = None
root.left.right = Node(7)

root.right.left = Node(10)
root.right.right = Node(6)

root.left.right.right = Node(10)
root.right.left.left = Node(6)

print(Solution().inOrder(root))
```

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def inOrder_iter(self, root):
        ans = []; stack = []; node = root
        while node or stack:
            while node: stack.append(node); node = node.left
            node = stack.pop(); ans.append(node.data); node = node.right
        return ans
```

```python
# root = [8, 1, 5, None, 7, 10, 6, None, 10, 6]
root = Node(8)
root.left = Node(1)
root.right = Node(5)

root.left.left = None
root.left.right = Node(7)

root.right.left = Node(10)
root.right.right = Node(6)

root.left.right.right = Node(10)
root.right.left.left = Node(6)

print(Solution().inOrder_iter(root))
```

## [Post-Order Traversal](https://www.geeksforgeeks.org/problems/postorder-traversal/1)

#### A postorder traversal first visits the left child (including its entire subtree), then visits the right child (including its entire subtree), and finally visits the node itself.

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def postOrder(self, root):
        ans = []
        def dfs(node):
            if node is None: return
            ans.append(node.data)
            dfs(node.right)
            dfs(node.left)
        dfs(root)
        return ans[::-1]
```

```python
# root = [19, 10, 8, 11, 13]
root = Node(19)
root.left = Node(10)
root.right = Node(8)

root.left.left = Node(11)
root.left.right = Node(13)

print(Solution().postOrder(root))
```

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def postOrder_iter(self, root):
        if root is None: return
        ans = []; stack = [root]
        while stack:
            node = stack.pop(); ans.append(node.data)
            if node.left: stack.append(node.left)
            if node.right: stack.append(node.right)
        return ans[::-1]
```

```python
# root = [19, 10, 8, 11, 13]
root = Node(19)
root.left = Node(10)
root.right = Node(8)

root.left.left = Node(11)
root.left.right = Node(13)

print(Solution().postOrder_iter(root))
```

## [Level Order Traversal (BFS)](https://www.geeksforgeeks.org/problems/level-order-traversal/1)

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

from queue import deque
class Solution:
    def levelOrder(self, root):
        q = deque([root]); ans = []
        while q:
            node = q.popleft(); ans.append(node.data)
            if node.left: q.append(node.left)
            if node.right: q.append(node.right)
        return ans
```

```python
# root = [10, 20, 30, 40, 50, N, N]
root = Node(10)
root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

print(Solution().levelOrder(root))
```

## [Pre-order, In-order and Post-Order in One Traversal](https://takeuforward.org/practice/dsa/pre%2C-post%2C-inorder-in-one-traversal)

| State | Action | Traversal |
|:---:|---|---|
| `1` | Add the node to `Pre` | Preorder |
| `2` | Add the node to `In` | Inorder |
| `3` | Add the node to `Post` | Postorder |

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def tree_traversal(self, root):
        Pre, In, Post = [], [], []; stack = [(root, 1)]
        while stack:
            node, state = stack.pop()
            if state == 1:
                Pre.append(node.data); stack.append((node, 2))
                if node.left: stack.append((node.left, 1))
            elif state == 2:
                In.append(node.data); stack.append((node, 3))
                if node.right: stack.append((node.right, 1))
            else: Post.append(node.data)
        return In, Pre, Post
```

```python
# root = [1, 3, 4, 5, 2, 7, 6 ]
root = Node(1)
root.left = Node(3)
root.right = Node(4)

root.left.left = Node(5)
root.right.right = Node(7)

root.left.right = Node(2)
root.right.left = Node(6)

print(*Solution().tree_traversal(root))
```

## [Height of a Binary Tree](https://www.geeksforgeeks.org/problems/height-of-binary-tree/1)

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

from queue import deque
class Solution:
    def height(self, root):
        q = deque([(root, 0)]); ans = 0
        while q:
            node, ht = q.popleft(); ans = max(ans, ht)
            if node.left: q.append((node.left, ht+1))
            if node.right: q.append((node.right, ht+1))
        return ans
```

```python
# root = [12, 8, 18, 5, 11]
root = Node(12)
root.left = Node(8)
root.right = Node(18)

root.left.left = Node(5)
root.left.right = Node(11)

print(Solution().height(root))
```

## [Balanced Tree Check](https://www.geeksforgeeks.org/problems/check-for-balanced-tree/1)

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.right = None
        self.left = None
        
from queue import deque
class Solution:
    def isBalanced(self, root):
        def height(node):
            if node is None: return 0
            left = height(node.left)
            if left == -1: return -1
            right = height(node.right)
            if right == -1: return -1
            if abs(left-right) > 1: return -1
            return 1 + max(left, right)
        return height(root) != -1
```

```python
# root = [10, 20, 30, 40, 60]
root = Node(10)
root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.right.right = Node(60)

print(Solution().isBalanced(root))
```

## [Diameter of Binary Tree](https://www.geeksforgeeks.org/problems/diameter-of-binary-tree/1)

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.right = None
        self.left = None

class Solution:
    def diameter(self, root):
        diam = 0
        def height(node):
            nonlocal diam
            if node is None: return 0
            left = height(node.left)
            right = height(node.right)
            diam = max(diam, left+right)
            return 1 + max(left, right)
        height(root)
        return diam
```

```python
# root = [1, 2, N, 3, 4, 5, N, 6]
root = Node(1)
root.left = Node(2)
root.right = Node(3)

root.left.left = Node(4)
root.left.right = Node(5)

root.right.right = Node(6)

print(Solution().diameter(root))
```

## [Maximum Path Sum](https://www.geeksforgeeks.org/problems/maximum-path-sum-from-any-node/1)

```python
class Node:
    def __init__(self,val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def findMaxSum(self, root): 
        sm = -10**11
        def dist(node):
            nonlocal sm
            if node is None: return 0
            left = dist(node.left)
            right = dist(node.right)
            sm = max(sm, node.data + left+right)
            return max(node.data + max(left, right, 0), 0)
        dist(root)
        return sm
```

```python
# root = [-80, -2, -100]
root = Node(-80)
root.left = Node(-2)
root.right = Node(-100)

print(Solution().findMaxSum(root))
```

## [Identical Trees](https://www.geeksforgeeks.org/problems/determine-if-two-trees-are-identical/1)

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.right = None
        self.left = None

class Solution:
    def isIdentical(self, r1, r2):
        if r1 is None and r2 is None: return True
        if r1 is None or r2 is None: return False
        if r1.data != r2.data: return False
        return (self.isIdentical(r1.left, r2.left) and self.isIdentical(r1.right, r2.right))
```

```python
# r1 = [1, 2, 3, 4], r2 = [1, 2, 3, N, N, 4]
r1 = Node(1)
r1.left = Node(2)
r1.right = Node(3)

r1.left.left = Node(4)

r2 = Node(1)
r2.left = Node(2)
r2.right = Node(3)

r2.right.right = Node(4)

print(Solution().isIdentical(r1, r2))
```

## [Zig-Zag Tree Traversal](https://www.geeksforgeeks.org/problems/zigzag-tree-traversal/1)

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

from queue import deque
class Solution:
    def zigZagTraversal(self, root):
        if root is None: return []
        q = deque([root]); ans = []; state = 0
        while q:
            level = []
            for _ in range(len(q)):
                node = q.popleft(); level.append(node.data)
                if node.left: q.append(node.left)
                if node.right: q.append(node.right)
            if state: level.reverse()
            ans.extend(level); state ^= 1
        return ans
```

```python
# root = [7, 9, 7, 8, 8, 6, N, 10, 9]
root = Node(7)
root.left = Node(9)
root.right = Node(7)

root.left.left = Node(8)
root.left.right = Node(8)

root.right.right = Node(6)

root.right.right.left = Node(10)
root.right.right.right = Node(9)

print(Solution().zigZagTraversal(root))
```

## [Tree Boundary Traversal](https://www.geeksforgeeks.org/problems/boundary-traversal-of-binary-tree/1)

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.right = None
        self.left = None

class Solution:
    def boundaryTraversal(self, root):
        if root is None: return []
        ans = []
        def isLeaf(node): return node.left is None and node.right is None
        def leftBoundary(node):
            while node:
                if not isLeaf(node): ans.append(node.data)
                if node.left: node = node.left
                else: node = node.right
        def leaves(node):
            if node is None: return
            if isLeaf(node): ans.append(node.data); return
            leaves(node.left); leaves(node.right)
        def rightBoundary(node):
            rb = []
            while node:
                if not isLeaf(node): rb.append(node.data)
                if node.right: node = node.right
                else: node = node.left
            ans.extend(rb[::-1])
        if not isLeaf(root): ans.append(root.data)
        leftBoundary(root.left); leaves(root); rightBoundary(root.right)
        return ans
```

```python
# root = [1, 2, 3, 4, 5, 6, 7, N, N, 8, 9, N, N, N, N]
root = Node(1)
root.left = Node(2)
root.right = Node(3)

root.left.left = Node(4)
root.left.right = Node(5)

root.right.left = Node(6)
root.right.right = Node(7)

root.left.right.left = Node(8)
root.left.right.right = Node(9)

print(Solution().boundaryTraversal(root))
```
