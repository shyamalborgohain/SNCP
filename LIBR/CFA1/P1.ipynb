```python
# https://codeforces.com/contest/2239/problem/D
MOD = 998244353

def init_comb(n):
    fac, ifac = [1]*(n+1), [1]*(n+1)
    for i in range(1, n+1): fac[i] = (fac[i-1]*i)%MOD
    ifac[n] = pow(fac[n], MOD-2, MOD)
    for i in range(n, 0, -1): ifac[i-1] = (ifac[i]*i)%MOD
    return fac, ifac

def C(n, r): return ((fac[n]*ifac[r])%MOD*ifac[n-r])%MOD

def sieve(n, k):
    g = [0]*(n+1); vis = [False]*(n+1); primes = []; g[1] = 1
    for i in range(2, n+1):
        if not vis[i]: primes.append(i); g[i] = pow(i, k, MOD)
        for p in primes:
            if i*p > n: break
            vis[i*p] = True; g[i*p] = (g[i]*g[p])%MOD
            if i%p == 0: break
    return g

import sys
f = open("Input/input7.txt", "rt")
sys.stdin = f

t = sys.stdin
for _ in range(int(t.readline())):
    n, m = map(int, t.readline().split())
    fac, ifac = init_comb(n)
    g = sieve(n, n-1); ans = 0
    for i in range(n-m+1):
        term = (g[n-i-1]*C(n-m, i))%MOD
        if i & 1: ans = (ans-term)%MOD
        else: ans = (ans+term)%MOD
    ans = (ans*((n-1)%MOD)*(C(n, m)%MOD))%MOD
    print(ans)
```

```python
# https://codeforces.com/problemset/problem/2194/E
import sys
INF = 10**31
f = open("Input/input14.txt", "rt")
sys.stdin = f

t = sys.stdin
for _ in range(int(t.readline())):
    n, m = map(int, t.readline().split()); a = [list(map(int, t.readline().split())) for _ in range(n)]
    dpS = [[-INF]*m for _ in range(n)]; dpS[0][0] = a[0][0]
    for i in range(n):
        for j in range(m):
            if i: dpS[i][j] = max(dpS[i][j], dpS[i-1][j] + a[i][j])
            if j: dpS[i][j] = max(dpS[i][j], dpS[i][j-1] + a[i][j])
    dpT = [[-INF]*m for _ in range(n)]; dpT[n-1][m-1] = a[n-1][m-1]
    for i in range(n-1, -1, -1):
        for j in range(m-1, -1, -1):
            if i+1 < n: dpT[i][j] = max(dpT[i][j], dpT[i+1][j] + a[i][j])
            if j+1 < m: dpT[i][j] = max(dpT[i][j], dpT[i][j+1] + a[i][j])
    routes = [[0]*m for _ in range(n)]; ans = dpS[n-1][m-1]
    for i in range(n):
        for j in range(m): routes[i][j] = dpS[i][j] + dpT[i][j] - a[i][j]
    for d in range(n+m-1):
        mx1 = -INF; mx2 = -INF              # largest and second largest route on the anti-diagonal
        r1 = max(0, d-m+1); r2 = min(n-1, d)
        for i in range(r1, r2+1):
            j = d-i; val = routes[i][j]
            if val > mx1: mx2 = mx1; mx1 = val
            elif val > mx2: mx2 = val
        for i in range(r1, r2+1):
            j = d-i; val = routes[i][j]
            avoid = mx2 if val == mx1 else mx1
            use = val - 2*a[i][j]
            ans = min(ans, max(use, avoid))
    print(ans)
```

```python
import sys
f = open("Input/input16.txt", "rt")
sys.stdin = f

t = sys.stdin
for _ in range(int(t.readline())):
    n = int(t.readline()); g = [[] for _ in range(n+1)]
    if n > 1:
        p = [0] + list(map(int, t.readline().split()))
        for i in range(1, n): g[p[i]].append(i+1); g[i+1].append(p[i])
    else: t.readline()
    leaf = [0] + list(map(int, t.readline().split())); k = max(leaf)
    if n==1: print("YES"); continue
    stack = [1]; order = []; par = [-1]*(n+1)
    while stack:
        u = stack.pop(); order.append(u)
        for v in g[u]:
            if v != par[u]: par[v] = u; stack.append(v)
    seg = {}; ok = True
    for u in reversed(order):
        if not ok: break
        if leaf[u]: seg[u] = (leaf[u], leaf[u]); continue
        segs = []
        for v in g[u]:
            if v != par[u]: segs.append(seg[v])
        sorted_segs = sorted(segs)
        for i in range(len(sorted_segs)-1):     # check consecutive segments
            if sorted_segs[i][1] + 1 != sorted_segs[i+1][0]: ok = False; break
        if not ok: break
        start = segs.index(sorted_segs[0]); m = len(segs)      # check cyclic shifts
        for i in range(m):
            if segs[(start+i)%m] != sorted_segs[i]: ok = False; break
        if not ok: break
        seg[u] = (sorted_segs[0][0], sorted_segs[-1][1])
    if ok and seg[1] == (1, k): print("YES")
    else: print("NO")
```

```python
import sys, math
f = open("Input/input17.txt", "rt")
sys.stdin = f

t = sys.stdin
for _ in range(int(t.readline())):
    n = int(t.readline()); a = list(map(int, t.readline().split())); g = [[] for _ in range(n)]
    for _ in range(n-1): x, y = map(int, t.readline().split()); x -= 1; y -= 1; g[x].append(y); g[y].append(x)
    ans = 0; ss = [1]*n; p = [-1]*n;order = []; stack = [0]; p[0] = -2
    while stack:
        u = stack.pop(); order.append(u)
        for v in g[u]:
            if v != p[u]: p[v] = u; stack.append(v)
    for u in reversed(order):
        val = []
        for v in g[u]:
            if v != p[u]: ss[u] += ss[v]; val.append(ss[v])
        val.append(n - ss[u])
        r = math.isqrt(a[u])
        if r*r < a[u]: continue
        total = 0; pairs = 0; triplets = 0
        for v in val: triplets += pairs*v; pairs += total*v; total += v
        ans += (triplets + pairs)
    print(ans)
```

```python
import sys
f = open("Input/input18.txt", "rt")
sys.stdin = f

t = sys.stdin
for _ in range(int(t.readline())):
    n, k1 = map(int, t.readline().split()); mxk = n*(n-1)//2; k = mxk - k1
    dp = [[False for _ in range(mxk+1)] for _ in range(n+1)]; p = [[-1 for _ in range(mxk+1)] for _ in range(n+1)]; dp[0][0] = True
    for i in range(n):
        for j in range(mxk+1):
            for l in range(1, n-i+1):
                if dp[i][j]: add = l*(l-1)//2; dp[i+l][j+add] = True; p[i+l][j+add] = l
    if not dp[n][k]: print(0); continue
    ans = []; segl = []; cur = n; curk = k; ln = n+1
    while cur > 0: l = p[cur][curk]; segl.append(l); cur -= l; curk -= l*(l-1)//2
    for s in segl: ans.extend(range(ln-s, ln)); ln -= s
    print(*ans)
```
