# Minimum Spanning Tree(MST)

## Prims Algorithm

### **Time Complexity:** $O((V + E) log V)$

```python
import heapq
class Solution:
    def spanningTree(self, V: int, edges: list[list[int]]) -> int:
        adj = [[] for _ in range(V)]
        for u,v,wt in edges: adj[u].append((v,wt)); adj[v].append((u,wt))
        pq = [(0,0,-1)]; visited = [False]*V; mn_wt = 0; mst = []
        while pq:
            wt,u,parent = heapq.heappop(pq)
            if visited[u] == True: continue
            visited[u] = True; mn_wt += wt
            if parent != -1: mst.append((u,parent))
            for v,wt in adj[u]:
                if not visited[v]: heapq.heappush(pq, (wt,v,u))
        return mn_wt, mst
```

```python
V = 3; edges = [[0, 1, 5], [1, 2, 3], [0, 2, 1]]
print(*Solution().spanningTree(V, edges))
```

## Disjoint-Set Union

For DSU with path compression + union by size/rank:
- find() → O(α(n)) amortized
- union() → O(α(n)) amortized
- Space → O(n)

Here α(n) is the inverse Ackermann function, which grows extremely slowly and is effectively constant for practical input sizes.

DSU operations are nearly O(1) amortized, specifically O(α(n)).

```python
parent = list(range(V))
rank = [0] * V

def find(x):
    if parent[x] != x: parent[x] = find(parent[x])
    return parent[x]

def union(a, b):
    a = find(a); b = find(b)
    if a == b: return False
    if rank[a] < rank[b]: a, b = b, a
    parent[b] = a
    if rank[a] == rank[b]: rank[a] += 1
    return True
```

## [MST-Kruskal's Algorithm](https://www.geeksforgeeks.org/problems/minimum-spanning-tree-kruskals-algorithm/1)

### **Time Complexity:** $O(E \log{E})$

### **Space Complexity:** $O(V+E)$

```python
from typing import List
class Solution:
    def kruskalsMST(self, V: int, edges: List[List[int]]) -> int:
        edges.sort(key=lambda x: x[2]); parent = list(range(V)); size = [1]*V
        def find(u):
            if parent[u] != u: parent[u] = find(parent[u])
            return parent[u]
        def union(a, b):
            a = find(a); b = find(b)
            if a == b: return False
            if size[a] < size[b]: a,b = b,a
            parent[b] = a; size[a] += size[b]
            return True
        mst = []; mst_wt = 0
        for u,v,wt in edges:
            if union(u,v): mst.append((u,v,wt)); mst_wt += wt
            if len(mst) == V-1: break
        return mst_wt, mst
```

```python
V = 10; edges = [[1, 4, 1], [1, 2, 2], [2, 3, 3], [2, 4, 3], [1, 5, 4], [3, 4, 5], [2, 6, 7], [3, 6, 8], [4, 5, 9]]
print(*Solution().kruskalsMST(V, edges))
```

## [Number of Provinces](https://www.geeksforgeeks.org/problems/number-of-provinces/1)

```python
class Solution:
    def countConnected(self, V, edges):
        parent = list(range(V))
        def find(u):
            if parent[u] != u: parent[u] = find(parent[u])
            return parent[u]
        def union(a, b):
            pa = find(a); pb = find(b)
            if pa != pb: parent[pa] = pb
        for u,v in edges: union(u,v)
        return len(set(find(i) for i in range(V)))
```

```python
V = 5; edges = [[0, 1], [2, 1], [3, 4]]
print(Solution().countConnected(V, edges))
```

## [Min Edge Movement to Connect a Graph](https://www.geeksforgeeks.org/problems/connecting-the-graph/1)

```python
class Solution:
    def minEdgesReq(self, n, edges):
        parent = list(range(n))
        if len(edges) < n-1: return -1
        def find(u):
            if parent[u] != u: parent[u] = find(parent[u])
            return parent[u]
        def union(a, b):
            pa = find(a); pb = find(b)
            if pa != pb: parent[pa] = pb
        for u,v in edges: union(u,v)
        return len(set(find(i) for i in range(n)))-1
```

```python
n = 10; edges = [[2,7], [6,1],[4,2], [3,2], [2,1], [6,8]]
print(Solution().minEdgesReq(n, edges))
```

## [Account Merge](https://www.geeksforgeeks.org/problems/account-merge/1)

```python
from collections import defaultdict
from typing import List
class Solution:
    def accMerge(self, arr: list[list[str]]) -> list[list[str]]:
        n = len(arr); parent = list(range(n)); size = [1]*n
        def find(u):
            if parent[u] != u: parent[u] = find(parent[u])
            return parent[u]
        def union(a,b):
            a = find(a); b = find(b)
            if a == b: return
            if size[a] < size[b]: a,b = b,a 
            parent[b] = a; size[a] += size[b]
        email_owner = {}
        for i, accounts in enumerate(arr):
            for email in accounts[1:]:
                if email in email_owner: union(i, email_owner[email])
                else: email_owner[email] = i
        merged = defaultdict(list); ans = []
        for email, owner in email_owner.items(): merged[find(owner)].append(email)
        for root, emails in merged.items(): emails.sort(); ans.append([arr[root][0]] + emails)
        return ans
```

```python
arr = [["John", "johnsmith@mail.com", "john_newyork@mail.com"], ["John", "johnsmith@mail.com", "john00@mail.com"], ["Mary", "mary@mail.com"], ["John", "johnnybravo@mail.com"]]
print(*Solution().accMerge(arr))
```

## [Count Islands After Every Operation](https://www.geeksforgeeks.org/problems/number-of-islands/1)

```python
from typing import List
class Solution:
    def numOfIslands(self, n: int, m: int, operators: List[List[int]]) -> List[int]:
        parent = list(range(n*m)); size = [1]*n*m; visited = [[False]*m for _ in range(n)]
        def find(u):
            if parent[u] != u: parent[u] = find(parent[u])
            return parent[u]
        def union(a, b):
            a = find(a); b = find(b)
            if a == b: return False
            if size[a] < size[b]: a,b = b,a
            parent[b] = a; size[a] += size[b]
            return True
        ans = []; dir = [(1,0),(-1,0),(0,1),(0,-1)]; islands = 0
        for x, y in operators:
            if visited[x][y]: ans.append(islands); continue
            visited[x][y] = True; islands += 1
            for dx, dy in dir:
                nx = x + dx; ny = y + dy
                if 0 <= nx < n and 0 <= ny < m and visited[nx][ny]:
                    a = m*x + y; b = m*nx + ny
                    if union(a, b): islands -= 1
            ans.append(islands)
        return ans
```

```python
n = 4; m = 5; operators = [[1,1],[0,1],[3,3],[3,4]]
print(Solution().numOfIslands(n, m, operators))
```

## [maximum Connected With One Change](https://www.geeksforgeeks.org/problems/maximum-connected-group/1)

```python
from typing import List
class Solution:
    def maxConnection(self, grid: list[list[int]]) -> int:
        n = len(grid); m = len(grid[0]); parent = list(range(n*m)); size = [1]*n*m
        def find(u):
            if parent[u] != u: parent[u] = find(parent[u])
            return parent[u]
        def union(a, b):
            a = find(a); b = find(b)
            if a == b: return False
            if size[a] < size[b]: a,b = b,a
            parent[b] = a; size[a] += size[b]
        dir = [(1,0),(-1,0),(0,1),(0,-1)]
        for x in range(n):
            for y in range(m):
                for dx, dy in dir:
                    if grid[x][y] == 0: continue
                    nx = x + dx; ny = y + dy
                    if 0 <= nx < n and 0 <= ny < m and grid[nx][ny] == 1: a = m*x + y; b = m*nx + ny; union(a, b)
        ans = max((size[find(x*m + y)] for x in range(n) for y in range(m) if grid[x][y] == 1), default=0)
        for x in range(n):
            for y in range(m):
                if grid[x][y] == 1: continue
                nbd = set(); val = 1
                for dx, dy in dir:
                    nx = x + dx; ny = y + dy
                    if 0 <= nx < n and 0 <= ny < m and grid[nx][ny] == 1: u = m*nx + ny; nbd.add(find(u))
                for root in nbd: val += size[root]
                ans = max(ans, val)
        return ans
        
```

```python
grid = [[1, 0, 1], [1, 0, 1], [1, 0, 1]]
print(Solution().maxConnection(grid))
```

## [Maximum Stone Removal](https://www.geeksforgeeks.org/problems/maximum-stone-removal-1662179442/1)

```python
class Solution:
    def maxRemove(self, stones: list[list[int]]) -> int:
        parent = list(range(n)); size = [1] * n
        def find(x):
            if parent[x] != x: parent[x] = find(parent[x])
            return parent[x]
        def union(a, b):
            a, b = find(a), find(b)
            if a == b: return
            if size[a] < size[b]: a, b = b, a
            parent[b] = a; size[a] += size[b]
        for i in range(n):
            for j in range(i):
                if stones[i][0] == stones[j][0] or stones[i][1] == stones[j][1]: union(i, j)
        components = len({find(i) for i in range(n)})
        return n - components
```

```python
stones = [[0, 0], [0, 1], [1, 0], [1, 2], [2, 1], [2, 2]]
print(Solution().maxRemove(stones))
```

## [Strongly Connected Components](https://www.geeksforgeeks.org/problems/strongly-connected-component-tarjanss-algo-1587115621/1)

```python
class Solution:
    def kosaraju(self, V: int, adj: list[list[int]]) -> list[list[int]]:
        visited = [False]*V; stack = []
        def dfs(u):
            visited[u] = True
            for v in adj[u]:
                if not visited[v]: dfs(v)
            stack.append(u)
        for u in range(V):
            if not visited[u]: dfs(u)
        rev = [[] for _ in range(V)]
        for u in range(V):
            for v in adj[u]: rev[v].append(u)
        visited = [False]*V; scc = []
        def rev_dfs(u, component):
            visited[u] = True; component.append(u)
            for v in rev[u]:
                if not visited[v]: rev_dfs(v, component)
        while stack:
            u = stack.pop()
            if not visited[u]:
                component = []; rev_dfs(u, component); scc.append(component)
        return scc
```

```python
V = 6; adj = [[1], [2], [0, 3], [4], [3, 5], []]
print(*Solution().kosaraju(V, adj))
```

## [Bridges in Network - Using Tarjan's Algorithm](https://leetcode.com/problems/critical-connections-in-a-network/description/)

```python
class Solution:
    def criticalConnections(self, n: int, connections: list[list[int]]) -> list[list[int]]:
        adj = [[] for _ in range(n)]
        for u,v in connections: adj[u].append(v); adj[v].append(u)
        tin = [-1]*n; low = [0]*n; bridges = []; timer = 0
        def dfs(u, parent):
            nonlocal timer; tin[u] = low[u] = timer; timer += 1
            for v in adj[u]:
                if v == parent: continue
                if tin[v] != -1: low[u] = min(low[u], tin[v])
                else:
                    dfs(v, u); low[u] = min(low[u], low[v])
                    if low[v] > tin[u]: bridges.append([u, v])
        for u in range(n):
            if tin[u] == -1: dfs(u, -1)
        return bridges
```

```python
n = 4; connections = [[0,1],[1,2],[2,0],[1,3]]
print(*Solution().criticalConnections(n, connections))
```

## [Articulation Points](https://www.geeksforgeeks.org/problems/articulation-point2616/1)

```python
class Solution:
    def articulationPoints(self, V: int, edges: list[list[int]]) -> list[int]:
        adj = [[] for _ in range(V)]
        for u,v in edges: adj[u].append(v); adj[v].append(u)
        tin = [-1]*V; low = [0]*V; visited = [False]*V; ap = [False]*V; timer = 0
        def dfs(u, parent):
            nonlocal timer; visited[u] = True; tin[u] = low[u] = timer; timer += 1; children = 0
            for v in adj[u]:
                if v == parent: continue
                if tin[v] != -1: low[u] = min(low[u], tin[v])
                else:
                    dfs(v, u); low[u] = min(low[u], low[v]); children += 1
                    if parent != -1 and low[v] >= tin[u]: ap[u] = True
                if parent == -1 and children > 1: ap[u] = True
        for u in range(V):
            if not visited[u]: dfs(u, -1)
        ans = [u for u in range(V) if ap[u]]
        return ans if ans else [-1]
```

```python
V = 5; edges = [[0, 1], [1, 4], [4, 3], [4, 2], [2, 3]]
print(*Solution().articulationPoints(V, edges))
```
