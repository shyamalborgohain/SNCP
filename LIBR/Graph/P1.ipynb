# **GRAPH REPRESENTATION**

## **Adjacency List**

${\large\text{Idea:}}$ Store all neighbors of each vertex.

${\large\text{Memory:}}$ O(V + E)

${\large\text{When to use:}}$ DFS, BFS, Trees, Sparse graphs, Most CP graph problems

${\large\text{Advantages:}}$ Memory efficient, Fast traversal

${\large\text{Disadvantages:}}$ Edge lookup is not O(1)

````python
n = 4
graph = [[] for _ in range(n + 1)]
edges = [(1, 2), (2, 3), (3, 4), (4, 1)]

for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)

print(*graph)
````

````python
# Traversal
node = 1
for nxt in graph[node]:
    print(nxt)    # prints vertices visited from node 1
````

## **Weighted Adjacency List**

${\large\text{Idea:}}$ Store(neighbor, weight)

${\large\text{Memory:}}$ O(V + E)

${\large\text{When to use:}}$ Dijkstra, Prim, Weighted Trees, Shortest Path Problems

${\large\text{Advantages:}}$ Efficient for weighted graphs

${\large\text{Disadvantages:}}$ Edge lookup is not O(1)

````python
n = 5
graph = [[] for _ in range(n + 1)]
edges = [(1, 2, 5), (1, 3, 2), (2, 4, 3), (3, 4, 4), (4, 5, 6)]

for u, v, w in edges:
    graph[u].append((v, w))
    graph[v].append((u, w))

print(*graph)
````

````python
# Traverse Neighbors of Node 4
node = 4
for nxt, wt in graph[node]:
    print(f"Neighbor = {nxt}, Weight = {wt}")
````

## **Edge List**

$\large{\text{Idea:}}$ Store only edges.

$\large{\text{Memory:}}$ O(E)

$\large{\text{When to use:}}$ Kruskal's MST, Sorting edges, Processing every edge(for counting degree of nodes)

$\large{\text{Advantages:}}$ Very simple, Easy to sort

$\large{\text{Disadvantages:}}$ DFS/BFS inefficient

````python
edges = [(1, 2), (1, 3), (2, 4), (3, 4), (3, 5)]
print(*edges)
````

````python
# Traversal
for u, v in edges:
    print(f"{u} -> {v}")
````

````python
# Count degree of each node
n = 5
degree = [0] * (n + 1)

for u, v in edges:
    degree[u] += 1
    degree[v] += 1

print(degree[1:])
````

````python
# Weighted Edges
edges = [(5, 1, 2), (2, 1, 3), (3, 2, 4), (4, 3, 4), (6, 3, 5)]   # Format (weight, u, v)

for w, u, v in edges:
    print(f"{u} -- {v} -weight {w}")
````

````python
# Sort by weight(Kruskal)
edges.sort()
for w, u, v in edges:
    print(w, u, v)
````

````python
# Find Total Weight
total = 0
for w, u, v in edges:
    total += w
print(total)
````

## **Adjacency Matrix**

$\large{\text{Idea:}}$ Store whether every pair of vertices is connected.

$\large{\text{Memory:}}$ O(V²)

$\large{\text{When to use:}}$ Dense graphs, Fast edge existence queries, Small n (typically ≤ 2000)

$\large{\text{Advantages:}}$ O(1) edge lookup

$\large{\text{Disadvantages:}}$ High memory usage

````python
n = 5
graph = [[0] * (n + 1) for _ in range(n + 1)]
edges = [(1, 2), (1, 3), (2, 4), (3, 4), (3, 5)]

for u, v in edges:
    graph[u][v] = 1
    graph[v][u] = 1

for row in graph[1:]:
    print(row[1:])
````

````python
# Traverse Neighbors of Entire Graph
for node in range(1, n + 1):
    lis = []
    for v in range(1, n + 1):
        if graph[node][v]:
            lis.append(v)
    print(f"Neighbors of {node}: {lis}")
````

## **Tree**

$\large{\text{Idea:}}$ A connected graph with n-1 edges.

$\large{\text{Storage:}}$ Adjacency List

$\large{\text{When to use:}}$ DFS, BFS, Tree DP, LCA, Subtree problems

````python
# Storage
n = 6
graph = [[] for _ in range(n + 1)]
edges = [(1, 2), (1, 3), (2, 4), (2, 5), (3, 6)]

for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)
print(*graph)
````

### DFS Traversal

````python
def dfs(u, parent):
    print(u)

    for v in graph[u]:
        if v != parent:
            dfs(v, u)

dfs(1, 0)
````

### Using Iterative DFS, find depth of each node

````python
depth = [0]*(n+1)
stack = [(1, 0)]

while stack:
    u, p = stack.pop()
    for v in graph[u]:
        if v != p:
            depth[v] = depth[u] + 1
            stack.append((v, u))

print(depth[1:])
````

### BFS Traversal

````python
# BFS
from collections import deque

q = deque([1])
visited = {1}

while q:
    u = q.popleft()
    print(u)

    for v in graph[u]:
        if v not in visited:
            visited.add(v)
            q.append(v)
````

### Rule of Thumb

- DFS/BFS → Adjacency List
- Trees → Adjacency List
- Weighted Graphs → Weighted Adjacency List
- Kruskal → Edge List
- Dense Graphs → Adjacency Matrix
- Need O(1) edge lookup → Adjacency Matrix
- Competitive Programming → Adjacency List (90% of the time)

### Complexity Cheat Sheet

| Representation | Memory | Edge Lookup | DFS/BFS |
|---------------|---------|-------------|---------|
| Edge List | O(E) | O(E) | ❌ |
| Adjacency List | O(V + E) | O(deg(v)) | ⭐ |
| Adjacency Matrix | O(V²) | O(1) | ✅ |
| Weighted Adj. List | O(V + E) | O(deg(v)) | ⭐ |

````python
# https://codeforces.com/problemset/problem/2114/E
import sys
# sys.setrecursionlimit(200000)  # This doesn't work, because it exceeds memory limit. You should use stack.
f = open("input1.txt", "rt")
sys.stdin = f

t = sys.stdin
for _ in range(int(t.readline())):
    n = int(t.readline()); vals = [0] + list(map(int, t.readline().split()))
    graph = [[] for _ in range(n+1)]
    for _ in range(n-1):
        u, v = map(int, t.readline().split())
        graph[u].append(v); graph[v].append(u)

    mx = [0]*(n+1)
    mn = [0]*(n+1)
    # dfs
    stack = [(1, 0)]
    while stack:
        u, p = stack.pop()
        if p == 0: mx[u] = mn[u] = vals[u]
        else: mx[u] = max(vals[u], vals[u] - mn[p]); mn[u] = min(vals[u], vals[u] - mx[p])
        for v in graph[u]:
            if v != p: stack.append((v, u))

    print(*mx[1:])
````

### Python DFS Notes

- Recursive DFS is fine for small depth.
- Default recursion limit ≈ `1000`.
- Deep trees can cause `RecursionError`.
- `sys.setrecursionlimit(...)` avoids `RecursionError`.
- Very deep recursion (e.g. `2e5`) may still cause **MLE** due to Python stack frames.
- For `n ≤ 2×10^5`, prefer **iterative DFS/BFS**.
